```python
# ============================================================
#  СТАТИСТИЧЕСКИЙ АНАЛИЗ ДЛЯ СПОРТИВНОЙ ПЕДАГОГИКИ
#
# ============================================================

# --- 1. УСТАНОВКА И ИМПОРТ БИБЛИОТЕК ---
!pip install pingouin openpyxl plotly -q

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
from scipy.stats import shapiro, levene, ttest_rel, ttest_ind, mannwhitneyu, wilcoxon
import pingouin as pg
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# Настройка стилей
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("viridis")
pd.set_option('display.max_columns', None)

print("✅ Библиотеки загружены!")
print("="*60)

# ============================================================
# 2. ЗАГРУЗКА ПОЛЬЗОВАТЕЛЬСКОГО ФАЙЛА
# ============================================================
print("\n ЗАГРУЗКА ДАННЫХ")
print("="*60)
print("""
ТРЕБОВАНИЯ К ФАЙЛУ:
1. Формат: .xlsx (Excel) или .csv
2. Файл должен содержать колонки:
   - 'Спортсмен' (или 'Имя', 'ФИО') — текстовые метки
   - 'Группа' (или 'Группа') — Экспериментальная / Контрольная
   - 'До' (или 'Before') — числовые значения до эксперимента
   - 'После' (или 'After') — числовые значения после эксперимента
3. Все колонки должны быть заполнены (без пустых ячеек)
""")

from google.colab import files
uploaded = files.upload()

# Определяем имя загруженного файла
file_name = list(uploaded.keys())[0]
print(f"\n📄 Загружен файл: {file_name}")

# Загружаем данные в зависимости от расширения
try:
    if file_name.endswith('.csv'):
        df = pd.read_csv(file_name)
    else:
        df = pd.read_excel(file_name)
    print("✅ Данные успешно загружены!")
except Exception as e:
    print(f"❌ Ошибка загрузки: {e}")
    print("Попробуйте сохранить файл в формате .xlsx или .csv и загрузить заново.")
    raise

# ============================================================
# 3. ПРЕДВАРИТЕЛЬНАЯ ОБРАБОТКА И ПРОВЕРКА ДАННЫХ
# ============================================================
print("\n" + "="*60)
print("ПРЕДВАРИТЕЛЬНАЯ ПРОВЕРКА ДАННЫХ")
print("="*60)

# Выводим первые строки для проверки
print("\n Первые 5 строк вашего файла:")
display(df.head())

# Проверяем названия колонок и переименовываем при необходимости
print("\n Проверка названий колонок...")

# Словарь для переименования (пользователь может добавить свои варианты)
rename_dict = {}
for col in df.columns:
    col_lower = col.lower().strip()
    if col_lower in ['спортсмен', 'имя', 'фио', 'участник', 'participant', 'name']:
        rename_dict[col] = 'Спортсмен'
    elif col_lower in ['группа', 'group', 'тип', 'type']:
        rename_dict[col] = 'Группа'
    elif col_lower in ['до', 'before', 'pre', 'начало', 'start']:
        rename_dict[col] = 'До'
    elif col_lower in ['после', 'after', 'post', 'конец', 'end']:
        rename_dict[col] = 'После'

if rename_dict:
    df = df.rename(columns=rename_dict)
    print(f"✅ Переименованы колонки: {rename_dict}")
else:
    print("⚠️ Переименование не требуется или колонки уже имеют правильные названия")

# Проверяем наличие обязательных колонок
required_cols = ['Спортсмен', 'Группа', 'До', 'После']
missing_cols = [col for col in required_cols if col not in df.columns]

if missing_cols:
    print(f"\n❌ ОШИБКА: Отсутствуют колонки: {missing_cols}")
    print(f"Доступные колонки: {list(df.columns)}")
    print("\nПожалуйста, переименуйте колонки в файле и загрузите заново.")
    print("Требуемые названия: 'Спортсмен', 'Группа', 'До', 'После'")
    raise ValueError("Отсутствуют обязательные колонки")

# Проверка на пустые значения
if df[required_cols].isnull().any().any():
    print("\n⚠️ ВНИМАНИЕ: Обнаружены пустые значения. Строки с пропусками будут удалены.")
    df = df.dropna(subset=required_cols)
    print(f"✅ После удаления пропусков: {len(df)} строк")

# Проверка уникальных групп
groups = df['Группа'].unique()
print(f"\n📌 Обнаружены группы: {list(groups)}")
if len(groups) != 2:
    print(f"⚠️ ВНИМАНИЕ: Ожидается 2 группы (Экспериментальная и Контрольная), найдено {len(groups)}")
    print("Убедитесь, что в колонке 'Группа' указаны корректные названия")

# Преобразуем числовые колонки
df['До'] = pd.to_numeric(df['До'], errors='coerce')
df['После'] = pd.to_numeric(df['После'], errors='coerce')

# Удаляем строки с некорректными числами
df = df.dropna(subset=['До', 'После'])

print(f"\n✅ Итоговое количество наблюдений: {len(df)}")

# Базовая статистика
print("\n Общая описательная статистика:")
display(df[['До', 'После']].describe().round(2))

# ============================================================
# 4. ФУНКЦИИ ДЛЯ АНАЛИЗА
# ============================================================

def interpret_effect(d):
    """Интерпретация величины эффекта Коэна"""
    if abs(d) < 0.2:
        return "Очень слабый (практически нет)"
    elif abs(d) < 0.5:
        return "Слабый эффект"
    elif abs(d) < 0.8:
        return "Средний эффект"
    else:
        return "СИЛЬНЫЙ эффект!"

def check_normality(data, group_col, value_col):
    """Проверка нормальности распределения"""
    groups = data[group_col].unique()
    results = []
    for group in groups:
        subset = data[data[group_col] == group][value_col]
        if len(subset) >= 3:
            stat, p = shapiro(subset)
            normal = "✅ ДА" if p > 0.05 else "❌ НЕТ"
            results.append([group, round(stat, 3), round(p, 4), normal])
        else:
            results.append([group, "N/A", "N/A", "⚠️ Малый размер выборки"])
    return pd.DataFrame(results, columns=['Группа', 'W-статистика', 'p-значение', 'Распределение нормальное?'])

def descriptive_stats(data, group_col, value_cols):
    """Расчет описательной статистики"""
    desc = data.groupby(group_col)[value_cols].agg(['mean', 'std', 'min', 'max', 'count'])
    desc.columns = ['Среднее', 'Ст.Откл.', 'Мин', 'Макс', 'N']
    return desc.round(2)

def compare_paired(before, after, group_name):
    """Сравнение ДО и ПОСЛЕ в одной группе"""
    diff = after - before
    mean_diff = np.mean(diff)
    std_diff = np.std(diff, ddof=1)

    # Проверка нормальности разностей
    if len(diff) >= 3:
        _, p_norm = shapiro(diff)
    else:
        p_norm = 0.01  # Маленькая выборка - используем непараметрику

    if p_norm > 0.05 and len(diff) >= 3:
        # Параметрический тест
        t_stat, p_val = ttest_rel(before, after)
        method = "Парный t-критерий Стьюдента"
        if std_diff != 0:
            cohen_d = abs(mean_diff / std_diff)
        else:
            cohen_d = 0
    else:
        # Непараметрический тест
        try:
            w_stat, p_val = wilcoxon(before, after)
            method = "Критерий Уилкоксона"
            n = len(before)
            mu = n * (n + 1) / 4
            sigma = np.sqrt(n * (n + 1) * (2*n + 1) / 24)
            z_val = (w_stat - mu) / sigma
            cohen_d = abs(z_val / np.sqrt(n))
            if mean_diff < 0:
                cohen_d = -cohen_d
        except:
            # Если тест не удался (например, все разности равны нулю)
            p_val = 1.0
            method = "Невозможно вычислить"
            cohen_d = 0

    return {
        'group': group_name,
        'mean_before': np.mean(before),
        'std_before': np.std(before),
        'mean_after': np.mean(after),
        'std_after': np.std(after),
        'mean_diff': mean_diff,
        'method': method,
        'p_value': p_val,
        'cohen_d': cohen_d,
        'significant': p_val < 0.05,
        'normality_p': p_norm
    }

def compare_independent(group1, group2):
    """Сравнение двух независимых групп"""
    # Проверка нормальности
    if len(group1) >= 3 and len(group2) >= 3:
        _, p_norm1 = shapiro(group1)
        _, p_norm2 = shapiro(group2)
        both_normal = (p_norm1 > 0.05) and (p_norm2 > 0.05)
    else:
        both_normal = False

    if both_normal:
        # Проверка равенства дисперсий
        try:
            _, p_levene = levene(group1, group2)
            if p_levene > 0.05:
                t_stat, p_val = ttest_ind(group1, group2, equal_var=True)
                method = "t-критерий (равные дисперсии)"
            else:
                t_stat, p_val = ttest_ind(group1, group2, equal_var=False)
                method = "t-критерий Уэлча (неравные дисперсии)"
        except:
            p_val = 1.0
            method = "Ошибка вычисления"
    else:
        try:
            u_stat, p_val = mannwhitneyu(group1, group2)
            method = "U-критерий Манна-Уитни"
            # Расчет эффекта через Z
            n1, n2 = len(group1), len(group2)
            mu_u = n1 * n2 / 2
            sigma_u = np.sqrt(n1 * n2 * (n1 + n2 + 1) / 12)
            z_val = (u_stat - mu_u) / sigma_u
            cohen_d = abs(z_val / np.sqrt(n1 + n2))
            return {
                'method': method,
                'mean1': np.mean(group1),
                'std1': np.std(group1),
                'mean2': np.mean(group2),
                'std2': np.std(group2),
                'p_value': p_val,
                'cohen_d': cohen_d,
                'significant': p_val < 0.05
            }
        except:
            p_val = 1.0
            method = "Ошибка вычисления"

    # Расчет Cohen's d для независимых
    n1, n2 = len(group1), len(group2)
    var1, var2 = np.var(group1, ddof=1), np.var(group2, ddof=1)
    pooled_std = np.sqrt(((n1-1)*var1 + (n2-1)*var2) / (n1+n2-2)) if (n1+n2-2) > 0 else 1
    cohen_d = abs(np.mean(group1) - np.mean(group2)) / pooled_std if pooled_std != 0 else 0

    return {
        'method': method,
        'mean1': np.mean(group1),
        'std1': np.std(group1),
        'mean2': np.mean(group2),
        'std2': np.std(group2),
        'p_value': p_val,
        'cohen_d': cohen_d,
        'significant': p_val < 0.05
    }

# ============================================================
# 5. ЭТАП 1: ОПИСАТЕЛЬНАЯ СТАТИСТИКА
# ============================================================
print("\n" + "="*60)
print("ЭТАП 1: ОПИСАТЕЛЬНАЯ СТАТИСТИКА")
print("="*60)

print("\n Описательная статистика ДО эксперимента:")
display(descriptive_stats(df, 'Группа', 'До'))

print("\n Описательная статистика ПОСЛЕ эксперимента:")
display(descriptive_stats(df, 'Группа', 'После'))

# Boxplot
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
sns.boxplot(x='Группа', y='До', data=df, palette='Set2', ax=axes[0])
axes[0].set_title('Распределение ДО эксперимента', fontsize=14)
axes[0].set_ylabel('Показатель')
sns.boxplot(x='Группа', y='После', data=df, palette='Set3', ax=axes[1])
axes[1].set_title('Распределение ПОСЛЕ эксперимента', fontsize=14)
axes[1].set_ylabel('Показатель')
plt.tight_layout()
plt.show()

# ============================================================
# 6. ЭТАП 2: ПРОВЕРКА НОРМАЛЬНОСТИ
# ============================================================
print("\n" + "="*60)
print("ЭТАП 2: ПРОВЕРКА НОРМАЛЬНОСТИ РАСПРЕДЕЛЕНИЯ")
print("="*60)

print("\n Для показателя 'До':")
display(check_normality(df, 'Группа', 'До'))

print("\n Для показателя 'После':")
display(check_normality(df, 'Группа', 'После'))

# Q-Q Plot
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
groups = df['Группа'].unique()
if len(groups) >= 2:
    titles = [
        f'До - {groups[0]}',
        f'До - {groups[1]}',
        f'После - {groups[0]}',
        f'После - {groups[1]}'
    ]
    axes = axes.flatten()
    plot_idx = 0
    for group in groups:
        for time_point in ['До', 'После']:
            subset = df[df['Группа'] == group][time_point]
            if len(subset) >= 3:
                stats.probplot(subset, dist="norm", plot=axes[plot_idx])
                axes[plot_idx].set_title(titles[plot_idx])
            else:
                axes[plot_idx].text(0.5, 0.5, f'Недостаточно данных\nдля {titles[plot_idx]}',
                                   ha='center', va='center', transform=axes[plot_idx].transAxes)
            plot_idx += 1
plt.tight_layout()
plt.show()

# ============================================================
# 7. ЭТАП 3: СРАВНЕНИЕ ДО И ПОСЛЕ
# ============================================================
print("\n" + "="*60)
print("ЭТАП 3: СРАВНЕНИЕ 'ДО' и 'ПОСЛЕ' В КАЖДОЙ ГРУППЕ")
print("="*60)

# Отделяем группы
group_names = df['Группа'].unique()
results_paired = {}

for group in group_names:
    before = df[df['Группа'] == group]['До']
    after = df[df['Группа'] == group]['После']
    if len(before) >= 3:
        results_paired[group] = compare_paired(before, after, group)
    else:
        print(f"⚠️ Группа '{group}' имеет слишком мало наблюдений ({len(before)}) для анализа")

# Вывод результатов
for group, res in results_paired.items():
    print(f"\n {res['group']} группа:")
    print(f"   Метод: {res['method']}")
    print(f"   ДО: {res['mean_before']:.2f} ± {res['std_before']:.2f}")
    print(f"   ПОСЛЕ: {res['mean_after']:.2f} ± {res['std_after']:.2f}")
    print(f"   Прирост: {res['mean_diff']:.2f}")
    print(f"   p-значение: {res['p_value']:.4f} {'✅ Значимо!' if res['significant'] else '❌ Не значимо'}")
    print(f"    Cohen's d = {res['cohen_d']:.3f} ({interpret_effect(res['cohen_d'])})")

# ============================================================
# 8. ЭТАП 4: СРАВНЕНИЕ ГРУПП МЕЖДУ СОБОЙ
# ============================================================
print("\n" + "="*60)
print("ЭТАП 4: СРАВНЕНИЕ ГРУПП МЕЖДУ СОБОЙ")
print("="*60)

if len(group_names) >= 2:
    group1 = df[df['Группа'] == group_names[0]]['После']
    group2 = df[df['Группа'] == group_names[1]]['После']

    if len(group1) >= 3 and len(group2) >= 3:
        res_ind = compare_independent(group1, group2)

        print(f"\n Сравнение групп ПОСЛЕ эксперимента:")
        print(f"   Метод: {res_ind['method']}")
        print(f"   {group_names[0]}: {res_ind['mean1']:.2f} ± {res_ind['std1']:.2f}")
        print(f"   {group_names[1]}: {res_ind['mean2']:.2f} ± {res_ind['std2']:.2f}")
        print(f"   Разница: {res_ind['mean1'] - res_ind['mean2']:.2f}")
        print(f"   p-значение: {res_ind['p_value']:.4f} {'✅ Значимо!' if res_ind['significant'] else '❌ Не значимо'}")
        print(f"    Cohen's d = {res_ind['cohen_d']:.3f} ({interpret_effect(res_ind['cohen_d'])})")
    else:
        print(f"⚠️ Недостаточно данных для сравнения групп")

# ============================================================
# 9. ЭТАП 5: КОРРЕЛЯЦИОННЫЙ АНАЛИЗ
# ============================================================
print("\n" + "="*60)
print("ЭТАП 5: КОРРЕЛЯЦИОННЫЙ АНАЛИЗ")
print("="*60)

if len(df) >= 3:
    _, p_norm_d = shapiro(df['До'])
    _, p_norm_a = shapiro(df['После'])

    if p_norm_d > 0.05 and p_norm_a > 0.05:
        corr_method = "Пирсон"
        corr_matrix = df[['До', 'После']].corr(method='pearson')
    else:
        corr_method = "Спирмен"
        corr_matrix = df[['До', 'После']].corr(method='spearman')

    print(f" Метод корреляции: {corr_method}")
    display(corr_matrix.round(3))

    # Scatter plot с линией регрессии
    fig, ax = plt.subplots(figsize=(10, 6))
    sns.regplot(x='До', y='После', data=df, scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
    ax.set_title(f'Корреляция показателей ({corr_method})', fontsize=14)
    r = corr_matrix.iloc[0, 1]
    ax.text(0.05, 0.95, f'R² = {r**2:.3f}', transform=ax.transAxes, fontsize=12,
            verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
    plt.grid(True, alpha=0.3)
    plt.show()

    print(f"\n Коэффициент корреляции: r = {r:.3f}")
    print(f" Коэффициент детерминации R² = {r**2:.3f}")
else:
    print("⚠️ Недостаточно данных для корреляционного анализа")

# ============================================================
# 10. ЭТАП 6: ИНТЕРАКТИВНЫЙ ГРАФИК
# ============================================================
print("\n" + "="*60)
print("ЭТАП 6: ИНДИВИДУАЛЬНАЯ ДИНАМИКА")
print("="*60)

fig_ind = go.Figure()

# Определяем цвета для групп
group_colors = {}
palette = ['#FF4444', '#4444FF', '#44AA44', '#FFAA00']
for i, group in enumerate(group_names):
    group_colors[group] = palette[i % len(palette)]

for i, row in df.iterrows():
    color = group_colors.get(row['Группа'], '#888888')
    fig_ind.add_trace(go.Scatter(
        x=['До', 'После'],
        y=[row['До'], row['После']],
        mode='lines+markers',
        name=row['Спортсмен'],
        line=dict(color=color, width=1.5),
        marker=dict(size=8),
        text=f"{row['Спортсмен']}<br>Группа: {row['Группа']}<br>Прирост: {row['После']-row['До']:.2f}",
        hoverinfo='text',
        showlegend=False
    ))

# Добавляем легенду для групп
for group in group_names:
    fig_ind.add_trace(go.Scatter(
        x=[None], y=[None],
        mode='markers',
        marker=dict(size=10, color=group_colors.get(group, '#888888')),
        name=group,
        showlegend=True
    ))

fig_ind.update_layout(
    title='Индивидуальная динамика каждого спортсмена',
    xaxis_title='Время тестирования',
    yaxis_title='Значение показателя',
    height=600,
    hovermode='closest',
    legend_title='Группа'
)
fig_ind.show()

# ============================================================
# 11. ДОПОЛНИТЕЛЬНАЯ ВИЗУАЛИЗАЦИЯ: СРАВНЕНИЕ ПРИРОСТОВ
# ============================================================
print("\n" + "="*60)
print("ДОПОЛНИТЕЛЬНАЯ ВИЗУАЛИЗАЦИЯ: ПРИРОСТЫ ГРУПП")
print("="*60)

df['Прирост'] = df['После'] - df['До']

fig_compare = go.Figure()
for group in group_names:
    subset = df[df['Группа'] == group]
    fig_compare.add_trace(go.Box(
        y=subset['Прирост'],
        name=group,
        boxmean='sd',
        marker_color=group_colors.get(group, '#888888'),
        boxpoints='all',
        jitter=0.3,
        pointpos=-1.8
    ))

fig_compare.update_layout(
    title='Сравнение приростов между группами',
    yaxis_title='Прирост показателя',
    height=500,
    hovermode='closest'
)
fig_compare.show()

# ============================================================
# 12. ИТОГОВОЕ ЗАКЛЮЧЕНИЕ
# ============================================================
print("\n" + "="*60)
print("ИТОГОВОЕ ЗАКЛЮЧЕНИЕ ПО РЕЗУЛЬТАТАМ ИССЛЕДОВАНИЯ")
print("="*60)

print(f"""
 АНАЛИТИЧЕСКИЙ ОТЧЕТ:

1. Объем выборки: N = {len(df)} спортсменов
2. Количество групп: {len(group_names)} ({', '.join(group_names)})
""")

for group, res in results_paired.items():
    print(f"""
3. Группа '{group}':
   - Прирост: {res['mean_diff']:.2f}
   - p = {res['p_value']:.4f} {'✅' if res['significant'] else '❌'}
   - Cohen's d = {res['cohen_d']:.3f} ({interpret_effect(res['cohen_d'])})
""")

if len(group_names) >= 2 and 'res_ind' in locals():
    print(f"""
4. Межгрупповое сравнение:
   - Разница между группами: {res_ind['mean1'] - res_ind['mean2']:.2f}
   - p = {res_ind['p_value']:.4f} {'✅' if res_ind['significant'] else '❌'}
   - Cohen's d = {res_ind['cohen_d']:.3f} ({interpret_effect(res_ind['cohen_d'])})
""")

if 'r' in locals():
    print(f"""
5. Корреляционная связь:
   - r = {r:.3f} (R² = {r**2:.3f})
""")

print("="*60)
print("\n АНАЛИЗ  ЗАВЕРШЕН!")
```

```python

"""
================================================================================
КОМПЛЕКСНЫЙ СТАТИСТИЧЕСКИЙ АНАЛИЗ ДЛЯ СПОРТИВНО-ПЕДАГОГИЧЕСКИХ ИССЛЕДОВАНИЙ
================================================================================

Включает: описательную статистику, проверку нормальности,
параметрические и непараметрические критерии, корреляционный,
регрессионный, дисперсионный, факторный и кластерный анализы,
повторные измерения, визуализацию.

================================================================================
"""

# ============================================================
# 0. ИМПОРТ БИБЛИОТЕК
# ============================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# Настройка графиков
plt.rcParams['figure.figsize'] = (14, 6)
plt.rcParams['font.size'] = 10
sns.set_theme(style="whitegrid", palette="muted")

# ============================================================
# 1. ГЕНЕРАЦИЯ СИНТЕТИЧЕСКИХ ДАННЫХ (спортивно-педагогическое исследование)
# ============================================================
np.random.seed(42)
n = 80  # общее число испытуемых

# Группы: контрольная и экспериментальная
group = np.array(['Экспериментальная'] * 40 + ['Контрольная'] * 40)

# Пол
gender = np.random.choice(['М', 'Ж'], size=n, p=[0.55, 0.45])

# Квалификация (качественная переменная)
qualification = np.random.choice(
    ['Начинающий', 'КМС', 'МС'], size=n, p=[0.3, 0.4, 0.3]
)

# азовые показатели (до эксперимента)
base_phys = np.random.normal(65, 10, n)   # физическая подготовленность
base_tech = np.random.normal(70, 8, n)    # техника
base_mot = np.random.normal(60, 12, n)    # мотивация

# После эксперимента: экспериментальная группа улучшается сильнее
post_phys = base_phys + np.where(
    group == 'Экспериментальная',
    np.random.normal(8, 3, n),
    np.random.normal(2, 3, n)
)
post_tech = base_tech + np.where(
    group == 'Экспериментальная',
    np.random.normal(7, 2.5, n),
    np.random.normal(1.5, 2.5, n)
)
post_mot = base_mot + np.where(
    group == 'Экспериментальная',
    np.random.normal(10, 4, n),
    np.random.normal(1, 4, n)
)

# Дополнительные показатели
coord = base_phys * 0.4 + base_tech * 0.3 + np.random.normal(0, 3, n)
speed = 100 - base_phys * 0.5 + np.random.normal(0, 4, n)

# Создание DataFrame
df = pd.DataFrame({
    'ID': range(1, n + 1),
    'Группа': group,
    'Пол': gender,
    'Квалификация': qualification,
    'Физ_подготовка_до': np.round(base_phys, 1),
    'Физ_подготовка_после': np.round(post_phys, 1),
    'Техника_до': np.round(base_tech, 1),
    'Техника_после': np.round(post_tech, 1),
    'Мотивация_до': np.round(base_mot, 1),
    'Мотивация_после': np.round(post_mot, 1),
    'Координация': np.round(coord, 1),
    'Быстрота': np.round(speed, 1),
})

# Динамика (разница)
df['Дин_физ'] = df['Физ_подготовка_после'] - df['Физ_подготовка_до']
df['Дин_тех'] = df['Техника_после'] - df['Техника_до']
df['Дин_мот'] = df['Мотивация_после'] - df['Мотивация_до']

print("=" * 70)
print("КОМПЛЕКСНЫЙ СТАТИСТИЧЕСКИЙ АНАЛИЗ")
print("=" * 70)
print("\n[1] ДАТАСЕТ СГЕНЕРИРОВАН")
print(f"    Общее число исптуемых: {df.shape[0]}")
print(f"    Группы: {df['Группа'].value_counts().to_dict()}")
print("\nПервые 5 строк:")
print(df.head())


# ============================================================
# 2. ОПИСАТЕЛЬНАЯ СТАТИСТИКА
# ============================================================
print("\n" + "=" * 70)
print("ЭТАП I. ОПИСАТЕЛЬНАЯ СТАТИСТИКА")
print("=" * 70)

numeric_cols = ['Физ_подготовка_до', 'Физ_подготовка_после',
                'Техника_до', 'Техника_после',
                'Мотивация_до', 'Мотивация_после',
                'Координация', 'Быстрота',
                'Дин_физ', 'Дин_тех', 'Дин_мот']

desc = df[numeric_cols].agg(['count', 'mean', 'median', 'std', 'min', 'max'])
desc.loc['std_error'] = df[numeric_cols].std() / np.sqrt(df[numeric_cols].count())
desc.loc['var'] = df[numeric_cols].var()
print("\n[2.1] Числовые характеристики выборки:")
print(np.round(desc, 2))

# --- График 1: Описательная статистика ---
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Гистограмма физической подготовки до и после
axes[0, 0].hist(df['Физ_подготовка_до'], bins=15, alpha=0.6, label='До', color='steelblue', edgecolor='black')
axes[0, 0].hist(df['Физ_подготовка_после'], bins=15, alpha=0.6, label='После', color='coral', edgecolor='black')
axes[0, 0].set_title('Распределение физической подготовленности (до / после)')
axes[0, 0].set_xlabel('Баллы')
axes[0, 0].set_ylabel('Частота')
axes[0, 0].legend()

# Boxplot по группам для динамики физ. пдготовки
sns.boxplot(data=df, x='Группа', y='Дин_физ', ax=axes[0, 1], palette='Set2')
axes[0, 1].set_title('Динамика физической подготовленности по группам')
axes[0, 1].set_ylabel('Прирост баллов')

# Boxplot мотивации до/после
mot_df = pd.melt(df, id_vars=['Группа'],
                 value_vars=['Мотивация_до', 'Мотивация_после'],
                 var_name='Этап', value_name='Мотивация')
mot_df['Этап'] = mot_df['Этап'].map({'Мотивация_до': 'До', 'Мотивация_после': 'После'})
sns.boxplot(data=mot_df, x='Группа', y='Мотивация', hue='Этап', ax=axes[1, 0], palette='coolwarm')
axes[1, 0].set_title('Мотивация: до и после эксперимента')

# Столбчатая диаграмма средних с ошибками
means = df.groupby('Группа')[['Дин_физ', 'Дин_тех', 'Дин_мот']].mean()
sems = df.groupby('Группа')[['Дин_физ', 'Дин_тех', 'Дин_мот']].sem()
means.T.plot(kind='bar', yerr=sems.T, ax=axes[1, 1], capsize=4, color=['#66c2a5', '#fc8d62'])
axes[1, 1].set_title('Средняя динамика показателей по группам (M ± SEM)')
axes[1, 1].set_ylabel('Прирост баллов')
axes[1, 1].set_xticklabels(['Дин. физ.', 'Дин. тех.', 'Дин. мот.'], rotation=0)
axes[1, 1].legend(title='Группа')

plt.tight_layout()
plt.savefig('fig1_descriptive_stats.png', dpi=150)
plt.show()
#print("\n    [График 1 сохранен: fig1_descriptive_stats.png]")


# ============================================================
# 3. ПРОВЕРКА НОРМАЛЬНОСТИ РАСПРЕДЕЛЕНИЯ
# ============================================================
print("\n" + "=" * 70)
print("ЭТАП II. ПРОВЕРКА НОРМАЛЬНОСТИ РАСПРЕДЕЛЕНИЯ")
print("=" * 70)

normality_results = []
for col in numeric_cols:
    # Shapiro-Wilk (n <= 50 берем подвыборку, иначе Kolmogorov-Smirnov)
    if len(df[col]) <= 50:
        stat, p = stats.shapiro(df[col])
        test_name = 'Shapiro-Wilk'
    else:
        # Lilliefors через Kolmogorov-Smirnov с оценкой параметров
        stat, p = stats.kstest((df[col] - df[col].mean()) / df[col].std(), 'norm')
        test_name = 'Kolmogorov-Smirnov'
    normality_results.append({
        'Показатель': col,
        'Тест': test_name,
        'W/KS': round(stat, 4),
        'p-value': round(p, 4),
        'Нормальность': 'Да' if p > 0.05 else 'Нет'
    })

norm_df = pd.DataFrame(normality_results)
print("\n[3.1] Результаты тестов на нормальность:")
print(norm_df.to_string(index=False))

# --- График 2: Q-Q plots ---
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
axes = axes.flatten()
for idx, col in enumerate(['Физ_подготовка_до', 'Физ_подготовка_после',
                           'Техника_до', 'Техника_после',
                           'Мотивация_до', 'Мотивация_после']):
    stats.probplot(df[col], dist="norm", plot=axes[idx])
    axes[idx].set_title(f'Q-Q plot: {col}')
    axes[idx].get_lines()[0].set_markerfacecolor('steelblue')
    axes[idx].get_lines()[1].set_color('coral')
plt.tight_layout()
plt.savefig('fig2_qqplots.png', dpi=150)
plt.show()
print("\n    [График 2 сохранен: fig2_qqplots.png]")


# ============================================================
# 4. ПРОВЕРКА ГОМОГЕННОСТИ ДИСПЕРСИЙ
# ============================================================
print("\n" + "=" * 70)
print("ЭТАП III. ПРОВЕРКА ГОМОГЕННОСТИ (ОДНОРОДНОСТИ) ДИСПЕРСИЙ")
print("=" * 70)

exp_group = df[df['Группа'] == 'Экспериментальная']
ctrl_group = df[df['Группа'] == 'Контрольная']

levene_vars = ['Физ_подготовка_до', 'Техника_до', 'Мотивация_до']
print("\n[4.1] Критерий Левене (2 группы):")
for var in levene_vars:
    stat, p = stats.levene(exp_group[var], ctrl_group[var])
    print(f"    {var}: Levene = {stat:.4f}, p = {p:.4f} -> {'Гомогенны' if p > 0.05 else 'Гетерогенны'}")

# --- Bartlett для >2 групп (по квалификации) ---
qual_groups = [group['Физ_подготовка_до'].values
               for name, group in df.groupby('Квалификация')]
stat, p = stats.bartlett(*qual_groups)
print(f"\n[4.2] Критерий Бартлетта (по квалификации):")
print(f"    Bartlett = {stat:.4f}, p = {p:.4f} -> {'Гомогенны' if p > 0.05 else 'Гетерогенны'}")


# ============================================================
# 5. СРАВНИТЕЛЬНЫЙ АНАЛИЗ ДВУХ ГРУПП
# ============================================================
print("\n" + "=" * 70)
print("ЭТАП IV. СРАВНИТЕЛЬНЫЙ АНАЛИЗ ДВУХ ГРУПП")
print("=" * 70)

comparison_vars = ['Физ_подготовка_до', 'Физ_подготовка_после',
                   'Техника_до', 'Техника_после',
                   'Мотивация_до', 'Мотивация_после',
                   'Дин_физ', 'Дин_тех', 'Дин_мот']

print("\n[5.1] Параметрический: t-критерий Стьюдента (независимые выборки)")
print("-" * 60)
ttest_results = []
for var in comparison_vars:
    t, p = stats.ttest_ind(exp_group[var], ctrl_group[var])
    # Cohen's d
    pooled_std = np.sqrt(((len(exp_group)-1)*exp_group[var].var() +
                          (len(ctrl_group)-1)*ctrl_group[var].var()) /
                         (len(exp_group)+len(ctrl_group)-2))
    cohens_d = (exp_group[var].mean() - ctrl_group[var].mean()) / pooled_std
    ttest_results.append({
        'Показатель': var,
        't': round(t, 3),
        'p': round(p, 4),
        'Cohen_d': round(cohens_d, 3),
        'Интерпретация_d': ('Малый' if abs(cohens_d) < 0.5 else
                            'Средний' if abs(cohens_d) < 0.8 else 'Большой')
    })
    print(f"    {var:25s}: t = {t:7.3f}, p = {p:.4f}, Cohen's d = {cohens_d:.3f}")

print("\n[5.2] Непараметрический: U-критерий Манна-Уитни")
print("-" * 60)
for var in comparison_vars:
    u, p = stats.mannwhitneyu(exp_group[var], ctrl_group[var], alternative='two-sided')
    print(f"    {var:25s}: U = {u:7.1f}, p = {p:.4f}")

# --- График 3: Сравнение групп ---
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
for idx, var in enumerate(['Дин_физ', 'Дин_тех', 'Дин_мот']):
    sns.violinplot(data=df, x='Группа', y=var, ax=axes[idx], palette='Set2', inner='box')
    axes[idx].set_title(f'{var}\n(t = {stats.ttest_ind(exp_group[var], ctrl_group[var])[0]:.2f}, '
                        f'p = {stats.ttest_ind(exp_group[var], ctrl_group[var])[1]:.4f})')
    axes[idx].set_ylabel('Прирост баллов')
plt.tight_layout()
plt.savefig('fig3_group_comparison.png', dpi=150)
plt.show()
print("\n    [График 3 сохранен: fig3_group_comparison.png]")


# ============================================================
# 6. ДИСПЕРСИОННЫЙ АНАЛИЗ (ANOVA)
# ============================================================
print("\n" + "=" * 70)
print("ЭТАП V. ДИСПЕРСИОННЫЙ АНАЛИЗ (ANOVA)")
print("=" * 70)

# Однофакторный ANOVA по квалификации (3 группы)
qual_grps = [grp['Дин_физ'].values for _, grp in df.groupby('Квалификация')]
f_stat, p_anova = stats.f_oneway(*qual_grps)
print("\n[6.1] Однофакторный ANOVA (динамика физ. подготовки по квалификации):")
print(f"    F = {f_stat:.4f}, p = {p_anova:.4f}")

# Ручной расчет eta-squared
ss_between = sum(len(g) * (np.mean(g) - df['Дин_физ'].mean())**2 for g in qual_grps)
ss_total = ((df['Дин_физ'] - df['Дин_физ'].mean())**2).sum()
eta_sq = ss_between / ss_total
print(f"    eta-squared = {eta_sq:.4f}")

# Попарные t-тесты с поправкой Бонферрони (вместо Tukey)
quals = df['Квалификация'].unique()
print("\n[6.2] Попарные сравнения (t-тест с поправкой Бонферрони):")
n_comparisons = len(quals) * (len(quals) - 1) // 2
alpha_corr = 0.05 / n_comparisons
for i in range(len(quals)):
    for j in range(i+1, len(quals)):
        g1 = df[df['Квалификация'] == quals[i]]['Дин_физ']
        g2 = df[df['Квалификация'] == quals[j]]['Дин_физ']
        t, p = stats.ttest_ind(g1, g2)
        print(f"    {quals[i]} vs {quals[j]}: t = {t:.3f}, p = {p:.4f} "
              f"({'*' if p < alpha_corr else 'нс'} при alpha = {alpha_corr:.4f})")

# --- График 4: ANOVA ---
fig, ax = plt.subplots(figsize=(10, 5))
sns.boxplot(data=df, x='Квалификация', y='Дин_физ', hue='Группа', palette='Set2', ax=ax)
ax.set_title(f'Динамика физ. подготовки по квалификации и группе\n(ANOVA: F = {f_stat:.2f}, p = {p_anova:.4f}, eta² = {eta_sq:.3f})')
plt.tight_layout()
plt.savefig('fig4_anova.png', dpi=150)
plt.show()
print("\n    [График 4 сохранен: fig4_anova.png]")


# ============================================================
# 7. КОРРЕЛЯЦИОННЫЙ АНАЛИЗ
# ============================================================
print("\n" + "=" * 70)
print("ЭТАП VI. КОРРЕЛЯЦИОННЫЙ АНАЛИЗ")
print("=" * 70)

corr_vars = ['Физ_подготовка_до', 'Техника_до', 'Мотивация_до', 'Координация', 'Быстрота']

# Пирсон
pearson_corr = df[corr_vars].corr(method='pearson')
print("\n[7.1] Матрица корреляций Пирсона (r):")
print(np.round(pearson_corr, 3))

# Спирмен
spearman_corr = df[corr_vars].corr(method='spearman')
print("\n[7.2] Матрица корреляций Спирмена (rho):")
print(np.round(spearman_corr, 3))

# Значимость отдельных корреляций
print("\n[7.3] Значимость корреляций (Пирсон):")
for i in range(len(corr_vars)):
    for j in range(i+1, len(corr_vars)):
        r, p = stats.pearsonr(df[corr_vars[i]], df[corr_vars[j]])
        print(f"    {corr_vars[i]} x {corr_vars[j]}: r = {r:.3f}, p = {p:.4f}")

# --- График 5: Корреляционная матрица ---
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
mask = np.triu(np.ones_like(pearson_corr, dtype=bool))
sns.heatmap(pearson_corr, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r',
            center=0, vmin=-1, vmax=1, ax=axes[0], square=True)
axes[0].set_title('Корреляции Пирсона')

sns.heatmap(spearman_corr, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r',
            center=0, vmin=-1, vmax=1, ax=axes[1], square=True)
axes[1].set_title('Корреляции Спирмена')
plt.tight_layout()
plt.savefig('fig5_correlation.png', dpi=150)
plt.show()
print("\n    [График 5 сохранен: fig5_correlation.png]")

# --- График 6: Scatter plots ---
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.regplot(data=df, x='Физ_подготовка_до', y='Координация', ax=axes[0],
            scatter_kws={'color': 'steelblue'}, line_kws={'color': 'coral'})
r, p = stats.pearsonr(df['Физ_подготовка_до'], df['Координация'])
axes[0].set_title(f'Физ. подготовка vs Координация\nr = {r:.3f}, p = {p:.4f}')

sns.regplot(data=df, x='Физ_подготовка_до', y='Быстрота', ax=axes[1],
            scatter_kws={'color': 'steelblue'}, line_kws={'color': 'coral'})
r, p = stats.pearsonr(df['Физ_подготовка_до'], df['Быстрота'])
axes[1].set_title(f'Физ. подготовка vs Быстрота\nr = {r:.3f}, p = {p:.4f}')
plt.tight_layout()
plt.savefig('fig6_scatter.png', dpi=150)
plt.show()
print("\n    [График 6 сохранен: fig6_scatter.png]")


# ============================================================
# 8. РЕГРЕССИОННЫЙ АНАЛИЗ
# ============================================================
print("\n" + "=" * 70)
print("ЭТАП VII. РЕГРЕССИОННЫЙ АНАЛИЗ")
print("=" * 70)

# Простая линейная регрессия: Координация ~ Физ_подготовка_до
X_simple = df[['Физ_подготовка_до']].values
y_simple = df['Координация'].values
model_simple = LinearRegression().fit(X_simple, y_simple)
r2_simple = model_simple.score(X_simple, y_simple)
print("\n[8.1] Простая линейная регрессия: Координация = f(Физ_подготовка_до)")
print(f"    Уравнение: Y = {model_simple.intercept_:.3f} + {model_simple.coef_[0]:.3f} * X")
print(f"    R-squared = {r2_simple:.4f}")

# F-тест для простой регрессии (через scipy)
n_obs = len(y_simple)
f_reg = (r2_simple / 1) / ((1 - r2_simple) / (n_obs - 2))
p_reg = 1 - stats.f.cdf(f_reg, 1, n_obs - 2)
print(f"    F = {f_reg:.4f}, p = {p_reg:.4f}")

# Множественная регрессия: Дин_физ ~ Физ_подготовка_до + Техника_до + Мотивация_до
X_multi = df[['Физ_подготовка_до', 'Техника_до', 'Мотивация_до']].values
y_multi = df['Дин_физ'].values
model_multi = LinearRegression().fit(X_multi, y_multi)
r2_multi = model_multi.score(X_multi, y_multi)
print("\n[8.2] Множественная регрессия: Дин_физ = f(Физ_подготовка_до, Техника_до, Мотивация_до)")
print(f"    Коэффициенты: {dict(zip(['Физ_до', 'Тех_до', 'Мот_до'], np.round(model_multi.coef_, 4)))}")
print(f"    Intercept = {model_multi.intercept_:.4f}")
print(f"    R-squared = {r2_multi:.4f}")

# --- График 7: Регрессия ---
fig, ax = plt.subplots(figsize=(10, 6))
y_pred_simple = model_simple.predict(X_simple)
ax.scatter(df['Физ_подготовка_до'], df['Координация'], alpha=0.6, color='steelblue', edgecolors='black')
ax.plot(df['Физ_подготовка_до'], y_pred_simple, color='coral', linewidth=2)
ax.set_xlabel('Физическая подготовка (до)')
ax.set_ylabel('Координация')
ax.set_title(f'Простая линейная регрессия\nY = {model_simple.intercept_:.2f} + {model_simple.coef_[0]:.2f}X, R² = {r2_simple:.3f}')
plt.tight_layout()
plt.savefig('fig7_regression.png', dpi=150)
plt.show()
print("\n    [График 7 сохране: fig7_regression.png]")


# ============================================================
# 9. АНАЛИЗ КАЧЕСТВЕННЫХ ПЕРЕМЕННЫХ (хи-квадрат)
# ============================================================
print("\n" + "=" * 70)
print("ЭТАП VIII. АНАЛИЗ КАЧЕСТВЕННЫХ ПЕРЕМЕННЫХ")
print("=" * 70)

# Таблица сопряженности: Группа x Квалификация
contingency = pd.crosstab(df['Группа'], df['Квалификация'])
print("\n[9.1] Таблица сопряженности (Группа x Квалификация):")
print(contingency)

chi2, p_chi, dof, expected = stats.chi2_contingency(contingency)
print(f"\n[9.2] Критерий хи-квадрат Пирсона:")
print(f"    chi2 = {chi2:.4f}, df = {dof}, p = {p_chi:.4f}")

# V Крамера
n_total = contingency.sum().sum()
cramers_v = np.sqrt(chi2 / (n_total * (min(contingency.shape) - 1)))
print(f"    V Крамера = {cramers_v:.4f}")

# --- График 8: Тепловая карта сопряженности ---
fig, ax = plt.subplots(figsize=(8, 4))
sns.heatmap(contingency, annot=True, fmt='d', cmap='YlOrRd', ax=ax)
ax.set_title(f'Таблица сопряженности\nchi2 = {chi2:.2f}, p = {p_chi:.4f}, V = {cramers_v:.3f}')
plt.tight_layout()
plt.savefig('fig8_chi2.png', dpi=150)
plt.show()
print("\n    [График 8 сохранен: fig8_chi2.png]")


# ============================================================
# 10. ПОВТОРНЫЕ ИЗМЕРЕНИЯ (парные критерии)
# ============================================================
print("\n" + "=" * 70)
print("ЭТАП IX. АНАЛИЗ ПОВТОРНЫХ ИЗМЕРЕНИЙ")
print("=" * 70)

# Парный t-критерий (до vs после) для экспериментальной группы
print("\n[10.1] Парный t-критерий (до vs после) — Экспериментальная группа:")
for before, after, name in [('Физ_подготовка_до', 'Физ_подготовка_после', 'Физ. подготовка'),
                             ('Техника_до', 'Техника_после', 'Техника'),
                             ('Мотивация_до', 'Мотивация_после', 'Мотивация')]:
    t, p = stats.ttest_rel(exp_group[before], exp_group[after])
    print(f"    {name}: t = {t:.3f}, p = {p:.4f}")

# T-критерий Вилкоксона (для контрольной группы как непараметрический аналог)
print("\n[10.2] T-критерий Вилкоксона (до vs после) — Контрольная группа:")
for before, after, name in [('Физ_подготовка_до', 'Физ_подготовка_после', 'Физ. подготовка'),
                             ('Техника_до', 'Техника_после', 'Техника'),
                             ('Мотивация_до', 'Мотивация_после', 'Мотивация')]:
    t, p = stats.wilcoxon(ctrl_group[before], ctrl_group[after])
    print(f"    {name}: T = {t:.1f}, p = {p:.4f}")

# --- График 9: Повторные измерения ---
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
measures = [('Физ_подготовка_до', 'Физ_подготовка_после', 'Физ. подготовка'),
            ('Техника_до', 'Техника_после', 'Техника'),
            ('Мотивация_до', 'Мотивация_после', 'Мотивация')]
for idx, (before, after, title) in enumerate(measures):
    for _, row in exp_group.iterrows():
        axes[idx].plot([0, 1], [row[before], row[after]], 'o-', color='steelblue', alpha=0.3)
    axes[idx].plot([0, 1],
                   [exp_group[before].mean(), exp_group[after].mean()],
                   'o-', color='red', linewidth=3, label='Среднее')
    axes[idx].set_xticks([0, 1])
    axes[idx].set_xticklabels(['До', 'После'])
    axes[idx].set_title(f'{title} (Эксп. группа)')
    axes[idx].legend()
plt.tight_layout()
plt.savefig('fig9_repeated_measures.png', dpi=150)
plt.show()
print("\n    [График 9 сохранен: fig9_repeated_measures.png]")


# ============================================================
# 11. ФАКТОРНЫЙ АНАЛИЗ (PCA как метод главных компонент)
# ============================================================
print("\n" + "=" * 70)
print("ЭТАП X. ФАКТРНЫЙ АНАЛИЗ (метод главных компонент)")
print("=" * 70)

fa_vars = ['Физ_подготовка_до', 'Техника_до', 'Мотивация_до', 'Координация', 'Быстрота']
X_fa = df[fa_vars].values
scaler = StandardScaler()
X_fa_scaled = scaler.fit_transform(X_fa)

pca = PCA(n_components=len(fa_vars))
pca.fit(X_fa_scaled)

print("\n[11.1] Объясненная дисперсия компонентами:")
for i, (var, ratio) in enumerate(zip(pca.explained_variance_, pca.explained_variance_ratio_)):
    print(f"    PC{i+1}: eigenvalue = {var:.3f}, объясненная дисперсия = {ratio:.3f} ({ratio*100:.1f}%)")

# Факторные нагрузки
loadings = pd.DataFrame(
    pca.components_[:3].T,
    columns=['PC1', 'PC2', 'PC3'],
    index=fa_vars
)
print("\n[11.2] Факторные нагрузки (первые 3 компоненты):")
print(np.round(loadings, 3))

# --- График 10: PCA ---
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].bar(range(1, len(pca.explained_variance_ratio_)+1),
            pca.explained_variance_ratio_, color='steelblue', edgecolor='black')
axes[0].set_xlabel('Главная компонента')
axes[0].set_ylabel('Доля объясненной дисперсии')
axes[0].set_title('Scree plot (PCA)')
axes[0].set_xticks(range(1, len(fa_vars)+1))

# Биплот
scores = pca.transform(X_fa_scaled)
axes[1].scatter(scores[:, 0], scores[:, 1], c=df['Группа'].map({'Экспериментальная': 'coral', 'Контрольная': 'steelblue'}), alpha=0.7, edgecolors='black')
for i, var in enumerate(fa_vars):
    axes[1].arrow(0, 0, pca.components_[0, i]*2, pca.components_[1, i]*2,
                  color='red', head_width=0.05)
    axes[1].text(pca.components_[0, i]*2.2, pca.components_[1, i]*2.2, var, color='red', fontsize=8)
axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
axes[1].set_title('Биплот PCA')
axes[1].legend(['Экспериментальная', 'Контрольная'])
plt.tight_layout()
plt.savefig('fig10_pca.png', dpi=150)
plt.show()
print("\n    [График 10 сохранен: fig10_pca.png]")


# ============================================================
# 12. КЛАСТЕРНЫЙ АНАЛИЗ
# ============================================================
print("\n" + "=" * 70)
print("ЭТАП XI. КЛАСТЕРНЫЙ АНАЛИЗ")
print("=" * 70)

# Иерархическая кластеризация
linked = linkage(X_fa_scaled, method='ward')
print("\n[12.1] Иерархическая кластеризация (метод Уорда):")
# K-means (k=3)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X_fa_scaled)
df['Кластер'] = kmeans.labels_
print(f"    K-means (k=3): распределение по кластерам = {pd.Series(kmeans.labels_).value_counts().to_dict()}")

# --- График 11: Кластерный анализ ---
fig, axes = plt.subplots(1, 2, figsize=(16, 5))
# Дендрограмма
dendrogram(linked, ax=axes[0], color_threshold=0.7*max(linked[:, 2]))
axes[0].set_title('Дендрограмма (метод Уорда)')
axes[0].set_xlabel('Испытуемые')
axes[0].set_ylabel('Евклидово расстояние')

# Кластеры на PCA
scatter = axes[1].scatter(scores[:, 0], scores[:, 1], c=kmeans.labels_, cmap='Set1', edgecolors='black', alpha=0.8)
axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
axes[1].set_title('Кластеры K-means (k=3) в пространстве PC1-PC2')
plt.colorbar(scatter, ax=axes[1], label='Кластер')
plt.tight_layout()
plt.savefig('fig11_cluster.png', dpi=150)
plt.show()
print("\n    [График 11 сохранен: fig11_cluster.png]")


# ============================================================
# 13. ИТОГОВАЯ СВОДКА
# ============================================================
print("\n" + "=" * 70)
print("ИТОГОВАЯ СВОДКА РЕЗУЛЬТАТОВ")
print("=" * 70)

print("""
1. ОПИСАТЕЛЬНАЯ СТАТИСТИКА:
   - Рассчитаны средние, медианы, стандартные отклонения и ошибки среднего
   - Визуализированы распределения и динамика по группам

2. ПРОВЕРКА НОРМАЛЬНОСТИ:
   - Тест Шапиро-Уилка / Колмогорова-Смирнова
   - Q-Q графики для визуальной оценки

3. ГОМОГЕННОСТЬ ДИСПЕРСИЙ:
   - Критерий Левене (2 группы)
   - Критерий Бартлетта (>2 групп)

4. СРАВНЕНИЕ ГРУПП:
   - t-критерий Стьюдента (параметрический)
   - U-критерий Манна-Уитни (непараметрический)
   - Размер эффекта: Cohen's d

5. ДИСПЕРСИОННЫЙ АНАЛИЗ:
   - Однофакторный ANOVA
   - eta-squared (размер эффекта)
   - Попарные сравнения с поправкой Бонферрони

6. КОРРЕЛЯЦИОННЫЙ АНАЛИЗ:
   - r Пирсона (количественные)
   - rho Спирмена (ранговые / без нормальности)
   - Матрицы корреляций + scatter plots

7. РЕГРЕССИОННЫЙ АНАЛИЗ:
   - Простая линейная регрессия
   - Множественная регрессия
   - R², F-тест

8. КАЧЕСТВЕННЫЕ ПЕРЕМЕННЫЕ:
   - Таблицы сопряженности
   - Критерий chi² Пирсона
   - V Крамера

9. ПОВТОРНЫЕ ИЗМЕРЕНИЯ:
   - Парный t-критерий
   - T-критерий Вилкоксона
   - Графики индивидуальных траекторий

10. ФАКТОРНЫЙ АНАЛИЗ:
    - PCA (метод главных компонент)
    - Scree plot, биплот, факторные нагрузки

11. КЛАСТЕРНЫЙ АНАЛИЗ:
    - Иерархическая кластеризация (дендрограмма)
    - K-means кластеризация
    - Визуализация кластеров в PCA-пространстве
""")
print("=" * 70)
print("АНАЛИЗ ЗАВЕРШЕН.")
print("=" * 70)
```

```python

"""
================================================================================
КОМПЛЕКСНЫЙ СТАТИСТИЧЕСКИЙ АНАЛИЗ

================================================================================


ИНСТРУКЦИЯ ПО НАСТРОЙКЕ:
1. Загрузите ваш CSV-файл в среду Colab (через панель слева или код)
2. В разделе "КОНФИГУРАЦИЯ" укажите:
   - путь к файлу
   - название столбца с группами
   - пары столбцов "до" и "после"
   - количественные показатели
   - качественные переменные
3. Запустите все ячейки
================================================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
import warnings
import os
warnings.filterwarnings('ignore')

plt.rcParams['figure.figsize'] = (14, 6)
plt.rcParams['font.size'] = 10
sns.set_theme(style="whitegrid", palette="muted")


# ============================================================================
# БЛОК ВСПОМОГАТЕЛЬНЫХ ФУНКЦИЙ
# ============================================================================

def robust_read_csv(filepath, **kwargs):
    """
    Надежная загрузка CSV с автоопределением кодировки, разделителя и наличия заголовка.
    Пробует несколько стратегий и выбирает ту, что сработала.
    """
    import io

    # Проверка существования файла
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"Файл не найден: {filepath}. "
                                f"Пожалуйста, загрузите файл в среду выполнения Colab.")

    # Сначала читаем сырые байты для анализа
    with open(filepath, 'rb') as f:
        raw_bytes = f.read()

    # Список кодировок для попытки (в порядке приоритета)
    encodings_to_try = []

    # Пробуем автоопределение через charset_normalizer (если доступен)
    try:
        from charset_normalizer import detect
        detection = detect(raw_bytes)
        detected_enc = detection.get('encoding')
        if detected_enc:
            encodings_to_try.append(detected_enc)
            print(f"[Загрузка] Автоопределена кодировка: {detected_enc} "
                  f"(уверенность: {detection.get('confidence', 'N/A')})")
    except Exception:
        pass

    # Добавляем стандартные кодировки
    standard_encodings = ['utf-8-sig', 'utf-8', 'cp1251', 'windows-1251',
                          'latin-1', 'iso-8859-1', 'cp1252', 'koi8-r']
    for enc in standard_encodings:
        if enc not in encodings_to_try:
            encodings_to_try.append(enc)

    # Возможные разделители
    delimiters = [',', ';', '\t', '|']

    # Пробуем все комбинации кодировка + разделитель
    errors = []
    for encoding in encodings_to_try:
        for sep in delimiters:
            try:
                # Пробуем прочитать с этой кодировкой и разделителем
                text = raw_bytes.decode(encoding)
                df_test = pd.read_csv(io.StringIO(text), sep=sep, **kwargs)

                # Проверяем, что получился непустой DataFrame
                if df_test.empty:
                    continue

                # Проверяем, что количество колонок разумное
                if len(df_test.columns) < 2:
                    continue

                # Проверяем, что нет NaN в названиях колонок (часто признак проблемы)
                if df_test.columns.isna().any():
                    continue

                print(f"[Загрузка] УСПЕХ: кодировка={encoding}, разделитель='{sep}', "
                      f"строк={len(df_test)}, столбцов={len(df_test.columns)}")
                print(f"[Загрузка] Столбцы: {list(df_test.columns)}")
                return df_test

            except Exception as e:
                errors.append(f"  {encoding} + '{sep}': {type(e).__name__}: {str(e)[:80]}")
                continue

    # Если ничего не сработало — выводим диагнстику и падаем
    print("[ОШИБКА] Не удалось прочитать CSV-файл ни с одной из комбинаций кодировка+разделитель.")
    print("\nПопробованные комбинации:")
    for err in errors[:15]:
        print(err)
    if len(errors) > 15:
        print(f"  ... и еще {len(errors) - 15} ошибок")
    raise ValueError("Не удалось загрузить CSV-файл. Проверьте формат файла.")


def interpret_p(p):
    """Развернутая интерпретация p-значения."""
    if p < 0.001:
        return (f"p < 0.001 — различия высоко достоверны. "
                f"Вероятность случайного получения таких различий менее 0.1%. "
                f"Нулевая гипотеза об отсутствии различий отвергается.")
    elif p < 0.01:
        return (f"p = {p:.4f} — различия достоверны н уровне значимости 1%. "
                f"Нулевая гипотеза отвергается.")
    elif p < 0.05:
        return (f"p = {p:.4f} — различия статистически значимы на уровне 5%. "
                f"Нулевая гипотеза отвергается.")
    else:
        return (f"p = {p:.4f} — различия статистически НЕзначимы. "
                f"Нет оснований отвергать нулевую гипотезу об отсутствии различий. "
                f"Полученные различия могут быть обусловлены случайными факторами.")


def interpret_cohens_d(d):
    """Интерпретация размера эффекта Коэна."""
    abs_d = abs(d)
    direction = "в пользу первой группы" if d > 0 else "в пользу второй группы" if d < 0 else ""
    if abs_d < 0.2:
        return f"Cohen's d = {d:.3f} — пренебрежимо малый размер эффекта {direction}"
    elif abs_d < 0.5:
        return f"Cohen's d = {d:.3f} — малый размер эффекта {direction}"
    elif abs_d < 0.8:
        return f"Cohen's d = {d:.3f} — средний размер эффекта {direction}"
    elif abs_d < 1.2:
        return f"Cohen's d = {d:.3f} — большой размер эффекта {direction}"
    else:
        return f"Cohen's d = {d:.3f} — очень большой размер эффекта {direction}"


def interpret_correlation(r, p, method="Пирсона"):
    """Интерпретация коэффициента корреляции."""
    abs_r = abs(r)
    if abs_r < 0.1:
        strength = "практически отсутствует"
    elif abs_r < 0.3:
        strength = "слабая"
    elif abs_r < 0.5:
        strength = "умеренная"
    elif abs_r < 0.7:
        strength = "заметная (средняя)"
    elif abs_r < 0.9:
        strength = "высокая"
    else:
        strength = "очень высокая"
    direction = "прямая" if r > 0 else "обратная"
    sig = "достоверна" if p < 0.05 else "НЕдостоверна"
    return (f"Коэффициент корреляции {method}: r = {r:.3f}. "
            f"Связь {direction}, {strength}. "
            f"Статистическая значимость: {sig} (p = {p:.4f}).")


def check_normality(data, alpha=0.05):
    """Проверка нормальности с автовыбором теста."""
    n = len(data)
    if n <= 50:
        stat, p = stats.shapiro(data)
        test_name = "Шапиро-Уилка"
    else:
        z = (data - np.mean(data)) / np.std(data, ddof=1)
        stat, p = stats.kstest(z, 'norm')
        test_name = "Колмогорова-Смирнова"
    is_normal = p > alpha
    return is_normal, p, test_name, stat


def auto_two_sample_test(group1, group2, paired=False, alpha=0.05):
    """Автоматический выбор и выполнение двухвыборочного теста."""
    g1, g2 = np.array(group1), np.array(group2)
    if paired:
        diff = g1 - g2
        is_normal, p_norm, test_name, stat_norm = check_normality(diff, alpha)
        if is_normal:
            stat, p = stats.ttest_rel(g1, g2)
            test_used = "парный t-критерий Стьюдента"
            interpretation = (f"Распределение разниц нормально ({test_name}: p = {p_norm:.4f}). "
                            f"Применен параметрический парный t-критерий.")
        else:
            stat, p = stats.wilcoxon(g1, g2)
            test_used = "T-критерий Вилкоксона"
            interpretation = (f"Распределение разниц отклоняется от нормального "
                            f"({test_name}: p = {p_norm:.4f}). Применен непараметрический "
                            f"T-критерий Вилкоксона.")
        return {
            'test': test_used,
            'statistic': stat,
            'p': p,
            'interpretation': interpretation,
            'normality_p': p_norm,
            'normality_test': test_name
        }
    else:
        n1, n2 = len(g1), len(g2)
        norm1, p1, t1, s1 = check_normality(g1, alpha)
        norm2, p2, t2, s2 = check_normality(g2, alpha)
        lev_stat, lev_p = stats.levene(g1, g2)
        equal_var = lev_p > alpha
        if norm1 and norm2:
            stat, p = stats.ttest_ind(g1, g2, equal_var=equal_var)
            test_used = "t-критерий Стьюдента" + (" (равные дисперсии)" if equal_var else " (неравные дисперсии, поправка Уэлча)")
            interpretation = (f"Обе выборки распределены нормально "
                            f"(группа 1: {t1} p = {p1:.4f}; группа 2: {t2} p = {p2:.4f}). "
                            f"Гомогенность диперсий: {'подтверждена' if equal_var else 'НЕ подтверждена'} "
                            f"(Levene p = {lev_p:.4f}). Применен {test_used}.")
            pooled_std = np.sqrt(((n1-1)*np.var(g1, ddof=1) + (n2-1)*np.var(g2, ddof=1)) / (n1+n2-2))
            cohens_d = (np.mean(g1) - np.mean(g2)) / pooled_std if pooled_std > 0 else 0
        else:
            stat, p = stats.mannwhitneyu(g1, g2, alternative='two-sided')
            test_used = "U-критерий Манна-Уитни"
            interpretation = (f"Хотя бы одна выборка НЕ распределена нормально "
                            f"(группа 1: {t1} p = {p1:.4f}; группа 2: {t2} p = {p2:.4f}). "
                            f"Применен непараметрический U-критерий Манна-Уитни.")
            cohens_d = None
        return {
            'test': test_used,
            'statistic': stat,
            'p': p,
            'interpretation': interpretation,
            'cohens_d': cohens_d,
            'normality': (norm1, norm2),
            'levene_p': lev_p,
            'equal_var': equal_var
        }


def interpret_anova(f_stat, p, eta_sq, group_names):
    """Интерпретация результатов ANOVA."""
    if p < 0.05:
        eta_interpret = (
            "пренебрежимо малый" if eta_sq < 0.01 else
            "малый" if eta_sq < 0.06 else
            "средний" if eta_sq < 0.14 else "большой"
        )
        return (f"ANOVA выявила статистически значимые различия между группами "
                f"({', '.join(group_names)}): F = {f_stat:.3f}, {interpret_p(p)} "
                f"Размер эффекта eta-squared = {eta_sq:.4f} ({eta_interpret}). "
                f"Это означает, что {eta_sq*100:.1f}% дисперсии зависимой переменной "
                f"объясняется принадлежностью к группам.")
    else:
        return (f"ANOVA НЕ выявила статистически значимых различий между группами "
                f"({', '.join(group_names)}): F = {f_stat:.3f}, {interpret_p(p)}")


# ============================================================================
# КОНФИГУРАЦИЯ — измените под ваши данные
# ============================================================================

from google.colab import files
uploaded = files.upload()

# Определяем имя загруженного файла
file_name = list(uploaded.keys())[0]
print(f"\n Загружен файл: {file_name}")

# Загружаем данные в зависимости от расширения
try:
    if file_name.endswith('.csv'):
        df = pd.read_csv(file_name)
    else:
        df = pd.read_excel(file_name)
    print("✅ Данные успешно загружены!")
except Exception as e:
    print(f"❌ Ошибка загрузки: {e}")
    print("Попробуйте сохранить файл в формате .xlsx или .csv и загрузить заново.")
    raise


# Путь к CSV-файлу. Если None — будут сгенерированы демо-данные.
# Примеры пуей для Colab:
#   CSV_PATH = '/content/my_data.csv'
#   CSV_PATH = '/content/data.csv'
CSV_PATH = '/content/report.csv'

# Столбец с группами (например, 'Группа', 'Group')
GROUP_COL = 'Группа'

# Список пар столбцов "до" и "после" для парного анализа
# Формат: [('название_до', 'название_после', 'человекочитаемое_название'), ...]
PAIRS_BEFORE_AFTER = [
    ('Физ_подготовка_до', 'Физ_подготовка_после', 'Физическая подготовленность'),
    ('Техника_до',        'Техника_после',        'Техника'),
    ('Мотивация_до',      'Мотивация_после',      'Мотивация'),
]

# Дополнительные количественные показатели (не входящие в пары до/после)
ADDITIONAL_NUMERIC = ['Координация', 'Быстрота']

# Качественные переменные для анализа таблиц сопряженности
CATEGORICAL_COLS = ['Пол', 'Квалификация']

# Зависимая переменная для регрессии
REGRESSION_TARGET = 'Дин_физ'

# Предикторы для множественной регрессии
REGRESSION_PREDICTORS = ['Физ_подготовка_до', 'Техника_до', 'Мотивация_до']

# Уровень значимости
ALPHA = 0.05


# ============================================================================
# 1. ЗАГРУЗКА ИЛИ ГЕНЕРАЦИЯ ДАННЫХ
# ============================================================================
print("=" * 75)
print(" КОМПЛЕКСНЫЙ СТАТИСТИЧЕСКИЙ АНАЛИЗ")
print("")
print("=" * 75)





if CSV_PATH:
    print(f"\n[1] ЗАГРУЗКА ДАННЫХ ИЗ: {CSV_PATH}")
    #df = robust_read_csv(CSV_PATH)
    print(f"\n    Данные успешно загружены: {df.shape[0]} строк x {df.shape[1]} столбцов")

    # Автоматически определяем числовые столбцы
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    if 'ID' in numeric_cols:
        numeric_cols.remove('ID')

    # Проверяем наличие указанных столбцов
    all_specified = []
    for before, after, _ in PAIRS_BEFORE_AFTER:
        all_specified.extend([before, after])
    all_specified.extend(ADDITIONAL_NUMERIC)
    all_specified.extend(CATEGORICAL_COLS)
    all_specified.append(GROUP_COL)

    missing = [c for c in all_specified if c not in df.columns]
    if missing:
        print(f"\n[ПРЕДУПРЕЖДЕНИЕ] Следующие столбцы из конфигурации не найдены в CSV: {missing}")
        print(f"    Доступные столбцы: {list(df.columns)}")
else:
    print("\n[1] CSV НЕ УКАЗАН — ГЕНЕРИРУЮТСЯ ДЕМОНСТРАЦИОННЫЕ ДАННЫЕ")
    np.random.seed(42)
    n = 80
    group = np.array(['Экспериментальная'] * 40 + ['Контрольная'] * 40)
    gender = np.random.choice(['М', 'Ж'], size=n, p=[0.55, 0.45])
    qualification = np.random.choice(['Начинающий', 'КМС', 'МС'], size=n, p=[0.3, 0.4, 0.3])
    base_phys = np.random.normal(65, 10, n)
    base_tech = np.random.normal(70, 8, n)
    base_mot = np.random.normal(60, 12, n)
    post_phys = base_phys + np.where(group == 'Экспериментальная', np.random.normal(8, 3, n), np.random.normal(2, 3, n))
    post_tech = base_tech + np.where(group == 'Экспериментальная', np.random.normal(7, 2.5, n), np.random.normal(1.5, 2.5, n))
    post_mot = base_mot + np.where(group == 'Экспериментальная', np.random.normal(10, 4, n), np.random.normal(1, 4, n))
    coord = base_phys * 0.4 + base_tech * 0.3 + np.random.normal(0, 3, n)
    speed = 100 - base_phys * 0.5 + np.random.normal(0, 4, n)
    df = pd.DataFrame({
        'ID': range(1, n+1),
        GROUP_COL: group,
        'Пол': gender,
        'Квалификация': qualification,
        'Физ_подготовка_до': np.round(base_phys, 1),
        'Физ_подготовка_после': np.round(post_phys, 1),
        'Техника_до': np.round(base_tech, 1),
        'Техника_после': np.round(post_tech, 1),
        'Мотивация_до': np.round(base_mot, 1),
        'Мотивация_после': np.round(post_mot, 1),
        'Координация': np.round(coord, 1),
        'Быстрота': np.round(speed, 1),
    })
    for before, after, _ in PAIRS_BEFORE_AFTER:
        df[f"Дин_{before.split('_')[0].lower()}"] = df[after] - df[before]

print(f"\nПервые 5 строк:")
print(df.head())
print(f"\nТипы данных:")
print(df.dtypes)


# ============================================================================
# 2. ОПИСАТЕЛЬНАЯ СТАТИСТИКА
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП I. ОПИСАТЕЛЬНАЯ СТАТИСТИКА")
print("=" * 75)

numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if 'ID' in numeric_cols:
    numeric_cols.remove('ID')

desc = df[numeric_cols].agg(['count', 'mean', 'median', 'std', 'min', 'max'])
desc.loc['std_error'] = df[numeric_cols].std() / np.sqrt(df[numeric_cols].count())
desc.loc['var'] = df[numeric_cols].var()
print("\n[2.1] Числовые характеристики выборки (M ± SD, медиана, мин-макс):")
print(np.round(desc.T, 2).to_string())

print("\n[2.2] ИНТЕРПРЕТАЦИЯ:")
for col in numeric_cols:
    mean_v = df[col].mean()
    std_v = df[col].std()
    cv = std_v / mean_v * 100 if mean_v != 0 else 0
    variability = (
        "низкая вариабельность" if cv < 10 else
        "средняя вариабельность" if cv < 20 else "высокая вариабельность"
    )
    print(f"    • {col}: M = {mean_v:.2f}, SD = {std_v:.2f}, CV = {cv:.1f}% — {variability}")

fig, axes = plt.subplots(2, min(len(PAIRS_BEFORE_AFTER), 3), figsize=(5*min(len(PAIRS_BEFORE_AFTER), 3), 10))
if len(PAIRS_BEFORE_AFTER) == 1:
    axes = np.array([[axes[0]], [axes[1]]])
for idx, (before, after, name) in enumerate(PAIRS_BEFORE_AFTER[:3]):
    axes[0, idx].hist(df[before], bins=15, alpha=0.6, label='До', color='steelblue', edgecolor='black')
    axes[0, idx].hist(df[after], bins=15, alpha=0.6, label='После', color='coral', edgecolor='black')
    axes[0, idx].set_title(f'{name}\n(распределение)')
    axes[0, idx].set_xlabel('Баллы')
    axes[0, idx].set_ylabel('Частота')
    axes[0, idx].legend()
    dyn_col = f"Дин_{before.split('_')[0].lower()}"
    if dyn_col in df.columns:
        sns.boxplot(data=df, x=GROUP_COL, y=dyn_col, ax=axes[1, idx], palette='Set2')
        axes[1, idx].set_title(f'Динамика: {name}')
        axes[1, idx].set_ylabel('Прирост')
plt.tight_layout()

plt.show()



# ============================================================================
# 3. ПРОВЕРКА НОРМАЛЬНОСТИ
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП II. ПРОВЕРКА НОРМАЛЬНОСТИ РАСПРЕДЕЛЕНИЯ")
print("=" * 75)

normality_results = []
print("\n[3.1] Результаты тестов на нормальность:")
for col in numeric_cols:
    is_normal, p, test_name, stat = check_normality(df[col].dropna(), ALPHA)
    normality_results.append({
        'Показатель': col,
        'Тест': test_name,
        'Статистика': round(stat, 4),
        'p-value': round(p, 4),
        'Нормальность': 'Да' if is_normal else 'Нет'
    })
    status = "РАСПРЕДЕЛЕНИЕ НОРМАЛЬНОЕ" if is_normal else "РАСПРЕДЕЛЕНИЕ ОТЛИЧАЕТСЯ ОТ НОРМАЛЬНОГО"
    print(f"    • {col}: {test_name} = {stat:.4f}, p = {p:.4f} — {status}")

norm_df = pd.DataFrame(normality_results)
print("\n[3.2] СВОДНАЯ ТАБЛИЦА:")
print(norm_df.to_string(index=False))

qq_cols = [before for before, _, _ in PAIRS_BEFORE_AFTER[:3]]
fig, axes = plt.subplots(1, len(qq_cols), figsize=(6*len(qq_cols), 5))
if len(qq_cols) == 1:
    axes = [axes]
for idx, col in enumerate(qq_cols):
    stats.probplot(df[col], dist="norm", plot=axes[idx])
    axes[idx].set_title(f'Q-Q plot: {col}')
    axes[idx].get_lines()[0].set_markerfacecolor('steelblue')
    axes[idx].get_lines()[1].set_color('coral')
plt.tight_layout()

plt.show()



# ============================================================================
# 4. ГОМОГЕННОСТЬ ДИСПЕРСИЙ
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП III. ПРОВЕРКА ГОМОГЕННОСТИ ДИСПЕРСИЙ")
print("=" * 75)

if GROUP_COL in df.columns and df[GROUP_COL].nunique() == 2:
    groups = df[GROUP_COL].unique()
    g1_name, g2_name = groups[0], groups[1]
    g1_data = df[df[GROUP_COL] == g1_name]
    g2_data = df[df[GROUP_COL] == g2_name]
    print(f"\n[4.1] Критерий Левене ля 2 групп ({g1_name} vs {g2_name}):")
    for col in [before for before, _, _ in PAIRS_BEFORE_AFTER]:
        stat, p = stats.levene(g1_data[col], g2_data[col])
        result = "ДИСПЕРСИИ ОДНОРОДНЫ" if p > ALPHA else "ДИСПЕРСИИ ГЕТЕРОГЕННЫ"
        print(f"    • {col}: Levene = {stat:.4f}, p = {p:.4f} — {result}")
        if p > ALPHA:
            print(f"      → Можно применять классический t-критерий Стьюдента с равными дисперсиями.")
        else:
            print(f"      → Следует использовать t-критерий с поправкой Уэлча (неравные дисперсии).")

if len(CATEGORICAL_COLS) > 0:
    for cat_col in CATEGORICAL_COLS:
        if df[cat_col].nunique() >= 3:
            print(f"\n[4.2] Критерий Бартлетта по '{cat_col}' (≥3 группы):")
            grps = [grp[PAIRS_BEFORE_AFTER[0][0]].values for _, grp in df.groupby(cat_col)]
            if all(len(g) > 1 for g in grps):
                stat, p = stats.bartlett(*grps)
                result = "ДИСПЕРСИИ ОДНОРОДНЫ" if p > ALPHA else "ДИСПЕРСИИ ГЕТЕРОГЕННЫ"
                print(f"    • {PAIRS_BEFORE_AFTER[0][2]}: Bartlett = {stat:.4f}, p = {p:.4f} — {result}")


# ============================================================================
# 5. СРАВНИТЕЛЬНЫЙ АНАЛИЗ ДВУХ ГРУПП
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП IV. СРАВНИТЕЛЬНЫЙ АНАЛИЗ ДВУХ ГРУПП")
print("=" * 75)

if GROUP_COL in df.columns and df[GROUP_COL].nunique() == 2:
    groups = df[GROUP_COL].unique()
    g1_name, g2_name = groups[0], groups[1]
    g1_data = df[df[GROUP_COL] == g1_name]
    g2_data = df[df[GROUP_COL] == g2_name]
    comparison_vars = [before for before, _, _ in PAIRS_BEFORE_AFTER] + \
                      [after for _, after, _ in PAIRS_BEFORE_AFTER] + \
                      [f"Дин_{before.split('_')[0].lower()}" for before, _, _ in PAIRS_BEFORE_AFTER] + \
                      ADDITIONAL_NUMERIC
    comparison_vars = [c for c in comparison_vars if c in df.columns]
    print(f"\n[5.1] СРАВНЕНИЕ ГРУПП '{g1_name}' (n={len(g1_data)}) и '{g2_name}' (n={len(g2_data)}):\n")
    comparison_results = []
    for col in comparison_vars:
        print(f"  ► ПОКАЗАТЕЛЬ: {col}")
        print(f"    Средние: {g1_name} = {g1_data[col].mean():.2f} ± {g1_data[col].std():.2f}; "
              f"{g2_name} = {g2_data[col].mean():.2f} ± {g2_data[col].std():.2f}")
        res = auto_two_sample_test(g1_data[col], g2_data[col], paired=False, alpha=ALPHA)
        print(f"    {res['interpretation']}")
        print(f"    Результат: {res['test']}: статистика = {res['statistic']:.3f}, {interpret_p(res['p'])}")
        if res['cohens_d'] is not None:
            print(f"    {interpret_cohens_d(res['cohens_d'])}")
        print()
        comparison_results.append({
            'Показатель': col,
            'Тест': res['test'],
            'Статистика': round(res['statistic'], 3),
            'p-value': round(res['p'], 4),
            'Cohen_d': round(res['cohens_d'], 3) if res['cohens_d'] is not None else None
        })
    dyn_cols = [c for c in comparison_vars if 'Дин_' in c]
    if len(dyn_cols) > 0:
        fig, axes = plt.subplots(1, len(dyn_cols), figsize=(6*len(dyn_cols), 5))
        if len(dyn_cols) == 1:
            axes = [axes]
        for idx, col in enumerate(dyn_cols):
            sns.violinplot(data=df, x=GROUP_COL, y=col, ax=axes[idx], palette='Set2', inner='box')
            axes[idx].set_title(col)
        plt.tight_layout()

        plt.show()

else:
    print(f"\n[5.1] ПРОПУЩЕНО: групповая переменная '{GROUP_COL}' не найдена или имеет ≠2 групп.")


# ============================================================================
# 6. ДИСПЕРСИОННЫЙ АНАЛИЗ (ANOVA)
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП V. ДИСПЕРСИОННЫЙ АНАЛИЗ (ANOVA)")
print("=" * 75)

anova_targets = []
if GROUP_COL in df.columns and df[GROUP_COL].nunique() > 2:
    anova_targets.append((GROUP_COL, df[GROUP_COL].unique()))
for cat_col in CATEGORICAL_COLS:
    if cat_col in df.columns and df[cat_col].nunique() >= 3:
        anova_targets.append((cat_col, df[cat_col].unique()))

for factor_col, factor_levels in anova_targets:
    print(f"\n[6.1] Однофакторный ANOVA по фактору '{factor_col}':")
    for before, after, name in PAIRS_BEFORE_AFTER:
        dyn_col = f"Дин_{before.split('_')[0].lower()}"
        target_col = dyn_col if dyn_col in df.columns else after
        grps = [grp[target_col].values for _, grp in df.groupby(factor_col)]
        if all(len(g) > 1 for g in grps):
            f_stat, p_anova = stats.f_oneway(*grps)
            ss_between = sum(len(g) * (np.mean(g) - df[target_col].mean())**2 for g in grps)
            ss_total = ((df[target_col] - df[target_col].mean())**2).sum()
            eta_sq = ss_between / ss_total if ss_total > 0 else 0
            print(f"\n  ► {name} (динамика):")
            print(f"    {interpret_anova(f_stat, p_anova, eta_sq, factor_levels)}")
            levels = df[factor_col].unique()
            n_comp = len(levels) * (len(levels) - 1) // 2
            alpha_corr = ALPHA / n_comp if n_comp > 0 else ALPHA
            print(f"    Попарные сравнения (поправка Бонферрони, alpha = {alpha_corr:.4f}):")
            for i in range(len(levels)):
                for j in range(i+1, len(levels)):
                    g1 = df[df[factor_col] == levels[i]][target_col]
                    g2 = df[df[factor_col] == levels[j]][target_col]
                    t, p = stats.ttest_ind(g1, g2)
                    sig = "*" if p < alpha_corr else "нс"
                    print(f"      {levels[i]} vs {levels[j]}: t = {t:.3f}, p = {p:.4f} ({sig})")

if len(anova_targets) > 0:
    factor_col = anova_targets[0][0]
    dyn_col = f"Дин_{PAIRS_BEFORE_AFTER[0][0].split('_')[0].lower()}"
    if dyn_col in df.columns:
        fig, ax = plt.subplots(figsize=(10, 5))
        hue = GROUP_COL if GROUP_COL in df.columns else None
        sns.boxplot(data=df, x=factor_col, y=dyn_col, hue=hue, palette='Set2', ax=ax)
        ax.set_title(f'Динамика по фактору "{factor_col}"')
        plt.tight_layout()

        plt.show()



# ============================================================================
# 7. ПАРНЫЙ АНАЛИЗ (до vs после)
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП VI. АНАЛИЗ ПОВТОРНЫХ ИЗМЕРЕНИЙ (до vs после)")
print("=" * 75)

print("\n[7.1] Сравнение показателей ДО и ПОСЛЕ эксперимента:\n")
for before, after, name in PAIRS_BEFORE_AFTER:
    print(f"  ► {name} ({before} vs {after}):")
    print(f"    Общая выборка: До = {df[before].mean():.2f} ± {df[before].std():.2f}; "
          f"После = {df[after].mean():.2f} ± {df[after].std():.2f}")
    res_all = auto_two_sample_test(df[before], df[after], paired=True, alpha=ALPHA)
    print(f"    {res_all['interpretation']}")
    print(f"    Результат: {res_all['test']}: статистика = {res_all['statistic']:.3f}, {interpret_p(res_all['p'])}")
    if GROUP_COL in df.columns:
        for grp_name in df[GROUP_COL].unique():
            grp_df = df[df[GROUP_COL] == grp_name]
            res_grp = auto_two_sample_test(grp_df[before], grp_df[after], paired=True, alpha=ALPHA)
            print(f"    • Группа '{grp_name}': {res_grp['test']}, статистика = {res_grp['statistic']:.3f}, "
                  f"p = {res_grp['p']:.4f}")
    print()

fig, axes = plt.subplots(1, len(PAIRS_BEFORE_AFTER), figsize=(6*len(PAIRS_BEFORE_AFTER), 5))
if len(PAIRS_BEFORE_AFTER) == 1:
    axes = [axes]
for idx, (before, after, name) in enumerate(PAIRS_BEFORE_AFTER):
    plot_df = pd.melt(df, id_vars=[GROUP_COL] if GROUP_COL in df.columns else ['ID'],
                      value_vars=[before, after], var_name='Этап', value_name=name)
    plot_df['Этап'] = plot_df['Этап'].map({before: 'До', after: 'После'})
    hue = GROUP_COL if GROUP_COL in df.columns else None
    sns.pointplot(data=plot_df, x='Этап', y=name, hue=hue, ax=axes[idx], palette='Set2', capsize=0.1)
    axes[idx].set_title(f'{name}\n(средние ± SEM)')
plt.tight_layout()

plt.show()



# ============================================================================
# 8. КОРРЕЛЯЦИОННЫЙ АНАЛИЗ
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП VII. КОРРЕЛЯЦИОННЫЙ АНАЛИЗ")
print("=" * 75)

corr_candidates = [before for before, _, _ in PAIRS_BEFORE_AFTER] + ADDITIONAL_NUMERIC
corr_vars = [c for c in corr_candidates if c in df.columns]

if len(corr_vars) >= 2:
    print("\n[8.1] Матрица корреляций Пирсона:")
    pearson_corr = df[corr_vars].corr(method='pearson')
    print(np.round(pearson_corr, 3).to_string())
    print("\n[8.2] Матрица корреляций Спирмена:")
    spearman_corr = df[corr_vars].corr(method='spearman')
    print(np.round(spearman_corr, 3).to_string())
    print("\n[8.3] ИНТЕРПРЕТАЦИЯ ЗНАЧИМЫХ КОРРЕЛЯЦИЙ (Пирсон):")
    for i in range(len(corr_vars)):
        for j in range(i+1, len(corr_vars)):
            r, p = stats.pearsonr(df[corr_vars[i]], df[corr_vars[j]])
            if p < ALPHA:
                print(f"    • {corr_vars[i]} x {corr_vars[j]}:")
                print(f"      {interpret_correlation(r, p, 'Пирсона')}")
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    mask = np.triu(np.ones_like(pearson_corr, dtype=bool))
    sns.heatmap(pearson_corr, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r',
                center=0, vmin=-1, vmax=1, ax=axes[0], square=True)
    axes[0].set_title('Корреляции Пирсона')
    sns.heatmap(spearman_corr, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r',
                center=0, vmin=-1, vmax=1, ax=axes[1], square=True)
    axes[1].set_title('Корреляции Спирмена')
    plt.tight_layout()

    plt.show()

    sig_pairs = []
    for i in range(len(corr_vars)):
        for j in range(i+1, len(corr_vars)):
            r, p = stats.pearsonr(df[corr_vars[i]], df[corr_vars[j]])
            if abs(r) > 0.3 and p < ALPHA:
                sig_pairs.append((corr_vars[i], corr_vars[j], r, p))
    if len(sig_pairs) > 0:
        n_sig = min(len(sig_pairs), 4)
        fig, axes = plt.subplots(1, n_sig, figsize=(5*n_sig, 5))
        if n_sig == 1:
            axes = [axes]
        for idx, (x, y, r, p) in enumerate(sig_pairs[:n_sig]):
            sns.regplot(data=df, x=x, y=y, ax=axes[idx],
                        scatter_kws={'color': 'steelblue'}, line_kws={'color': 'coral'})
            axes[idx].set_title(f'{x} vs {y}\nr = {r:.3f}, p = {p:.4f}')
        plt.tight_layout()

        plt.show()



# ============================================================================
# 9. РЕГРЕССИОННЫЙ АНАЛИЗ
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП VIII. РЕГРЕССИОННЫЙ АНАЛИЗ")
print("=" * 75)

if len(corr_vars) >= 2:
    x_col, y_col = corr_vars[0], corr_vars[1]
    X_s = df[[x_col]].values
    y_s = df[y_col].values
    model_s = LinearRegression().fit(X_s, y_s)
    r2_s = model_s.score(X_s, y_s)
    n_s = len(y_s)
    f_reg_s = (r2_s / 1) / ((1 - r2_s) / (n_s - 2)) if r2_s < 1 else np.inf
    p_reg_s = 1 - stats.f.cdf(f_reg_s, 1, n_s - 2) if f_reg_s != np.inf else 0.0
    print(f"\n[9.1] Простая линейная регрессия: {y_col} = f({x_col})")
    print(f"    Уравнение: Y = {model_s.intercept_:.3f} + {model_s.coef_[0]:.3f} * X")
    print(f"    R-squared = {r2_s:.4f} — модель объясняет {r2_s*100:.1f}% дисперсии зависимой переменной.")
    print(f"    F = {f_reg_s:.3f}, p = {p_reg_s:.4f}")
    if p_reg_s < ALPHA:
        print(f"    → Регрессионная модель статистически значима. "
              f"Прдиктор '{x_col}' достоверно влияет на '{y_col}'.")
    else:
        print(f"    → Регрессионная модель статистически НЕзначима.")

preds = [p for p in REGRESSION_PREDICTORS if p in df.columns]
if REGRESSION_TARGET in df.columns and len(preds) > 0:
    X_m = df[preds].values
    y_m = df[REGRESSION_TARGET].values
    model_m = LinearRegression().fit(X_m, y_m)
    r2_m = model_m.score(X_m, y_m)
    n_m, k_m = len(y_m), len(preds)
    f_reg_m = (r2_m / k_m) / ((1 - r2_m) / (n_m - k_m - 1)) if r2_m < 1 else np.inf
    p_reg_m = 1 - stats.f.cdf(f_reg_m, k_m, n_m - k_m - 1) if f_reg_m != np.inf else 0.0
    print(f"\n[9.2] Множественная регрессия: {REGRESSION_TARGET} = f({', '.join(preds)})")
    print(f"    Коэффициенты:")
    for name, coef in zip(preds, model_m.coef_):
        direction = "положительное" if coef > 0 else "отрицательное"
        print(f"      • {name}: {coef:.4f} ({direction} влияние)")
    print(f"    Intercept = {model_m.intercept_:.4f}")
    print(f"    R-squared = {r2_m:.4f} — совокупность предикторов объясняет {r2_m*100:.1f}% дисперсии.")
    print(f"    F = {f_reg_m:.3f}, p = {p_reg_m:.4f}")
    if p_reg_m < ALPHA:
        print(f"    → Множественная регрессионная модель в целом статистически значима.")
    else:
        print(f"    → Множественная регрессионная модель в целом статистически НЕзначима.")
    print(f"    Интерпретация: при изменении предикторов на 1 балл результат изменяется "
          f"согласно коэффициентам регрессии.")

fig, ax = plt.subplots(figsize=(10, 6))
y_pred_s = model_s.predict(X_s)
ax.scatter(df[x_col], df[y_col], alpha=0.6, color='steelblue', edgecolors='black')
ax.plot(df[x_col], y_pred_s, color='coral', linewidth=2)
ax.set_xlabel(x_col)
ax.set_ylabel(y_col)
ax.set_title(f'Простая линейная регрессия\nY = {model_s.intercept_:.2f} + {model_s.coef_[0]:.2f}X, R² = {r2_s:.3f}')
plt.tight_layout()

plt.show()



# ============================================================================
# 10. КАЧЕСТВЕННЫЕ ПЕРЕМЕННЫЕ (хи-квадрат)
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП IX. АНАЛИЗ КАЧЕСТВЕННЫХ ПЕРЕМЕННЫХ")
print("=" * 75)

if GROUP_COL in df.columns and len(CATEGORICAL_COLS) > 0:
    for cat_col in CATEGORICAL_COLS:
        if cat_col in df.columns:
            contingency = pd.crosstab(df[GROUP_COL], df[cat_col])
            print(f"\n[10.1] Таблица сопряженности: '{GROUP_COL}' x '{cat_col}':")
            print(contingency)
            chi2, p_chi, dof, expected = stats.chi2_contingency(contingency)
            n_total = contingency.sum().sum()
            cramers_v = np.sqrt(chi2 / (n_total * (min(contingency.shape) - 1))) if min(contingency.shape) > 1 else 0
            print(f"\n[10.2] Критерий хи-квадрат:")
            print(f"    chi2 = {chi2:.4f}, df = {dof}, p = {p_chi:.4f}")
            print(f"    {interpret_p(p_chi)}")
            v_interp = (
                "отсутствует" if cramers_v < 0.1 else
                "слабая" if cramers_v < 0.3 else
                "умеренная" if cramers_v < 0.5 else "заметная"
            )
            print(f"    V Крамера = {cramers_v:.4f} — связь {v_interp}.")
            if p_chi < ALPHA:
                print(f"    → Между группой и '{cat_col}' существует статистически значимая связь.")
            else:
                print(f"    → Связь между группой и '{cat_col}' статистически НЕзначима.")
    if len(CATEGORICAL_COLS) > 0:
        fig, ax = plt.subplots(figsize=(8, 4))
        cat_col = CATEGORICAL_COLS[0]
        contingency = pd.crosstab(df[GROUP_COL], df[cat_col])
        sns.heatmap(contingency, annot=True, fmt='d', cmap='YlOrRd', ax=ax)
        ax.set_title(f'Таблица сопряженности: {GROUP_COL} x {cat_col}')
        plt.tight_layout()

        plt.show()



# ============================================================================
# 11. ФАКТОРНЫЙ АНАЛИЗ (PCA)
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП X. ФАКТОРНЫЙ АНАЛИЗ (метод главных компонент)")
print("=" * 75)

fa_vars = [c for c in corr_vars if c in df.columns]
if len(fa_vars) >= 3:
    X_fa = df[fa_vars].dropna().values
    scaler = StandardScaler()
    X_fa_scaled = scaler.fit_transform(X_fa)
    n_comp = min(len(fa_vars), 5)
    pca = PCA(n_components=n_comp)
    pca.fit(X_fa_scaled)
    print("\n[11.1] Объясненная дисперсия компонентами:")
    for i, (var, ratio) in enumerate(zip(pca.explained_variance_, pca.explained_variance_ratio_)):
        print(f"    PC{i+1}: eigenvalue = {var:.3f}, доля дисперсии = {ratio:.3f} ({ratio*100:.1f}%)")
    cumulative = np.cumsum(pca.explained_variance_ratio_)
    n_80 = np.argmax(cumulative >= 0.80) + 1
    print(f"\n    Для объяснения 80% дисперсии достаточно {n_80} компонент().")
    print(f"    Общая объясненная дисперсия первых 3 компонент: {cumulative[min(2, n_comp-1)]*100:.1f}%")
    loadings = pd.DataFrame(
        pca.components_[:3].T,
        columns=['PC1', 'PC2', 'PC3'],
        index=fa_vars
    )
    print("\n[11.2] Факторные нагрузки (первые 3 компоненты):")
    print(np.round(loadings, 3).to_string())
    print("\n    Интерпретация факторных нагрузок:")
    for var in fa_vars:
        pc_loads = loadings.loc[var]
        max_pc = pc_loads.abs().idxmax()
        max_val = pc_loads[max_pc]
        print(f"    • {var}: наибольшая нагрузка на {max_pc} = {max_val:.3f} "
              f"({'сильная' if abs(max_val) > 0.7 else 'умеренная' if abs(max_val) > 0.4 else 'слабая'} связь)")
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    axes[0].bar(range(1, n_comp+1), pca.explained_variance_ratio_, color='steelblue', edgecolor='black')
    axes[0].set_xlabel('Главная компонента')
    axes[0].set_ylabel('Доля объясненной дисперсии')
    axes[0].set_title('Scree plot')
    axes[0].set_xticks(range(1, n_comp+1))
    scores = pca.transform(X_fa_scaled)
    group_vals = df[GROUP_COL].values if GROUP_COL in df.columns else np.array([''] * len(scores))
    colors = {'Экспериментальная': 'coral', 'Контрольная': 'steelblue'}
    for grp in np.unique(group_vals):
        mask = group_vals == grp
        c = colors.get(grp, 'gray')
        axes[1].scatter(scores[mask, 0], scores[mask, 1], c=c, label=grp, alpha=0.7, edgecolors='black')
    for i, var in enumerate(fa_vars):
        axes[1].arrow(0, 0, pca.components_[0, i]*2.5, pca.components_[1, i]*2.5,
                      color='red', head_width=0.05)
        axes[1].text(pca.components_[0, i]*2.7, pca.components_[1, i]*2.7, var, color='red', fontsize=8)
    axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
    axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
    axes[1].set_title('Биплот PCA')
    axes[1].legend()
    plt.tight_layout()

    plt.show()



# ============================================================================
# 12. КЛАСТЕРНЫЙ АНАЛИЗ
# ============================================================================
print("\n" + "=" * 75)
print("ЭТАП XI. КЛАСТЕРНЫЙ АНАЛИЗ")
print("=" * 75)

if len(fa_vars) >= 3:
    linked = linkage(X_fa_scaled, method='ward')
    kmeans = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X_fa_scaled)
    df['Кластер'] = kmeans.labels_
    print("\n[12.1] Иерархическая кластеризация (метод Уорда):")
    print("    Построена дендрограмма для визуальной оценки числа кластеров.")
    print("\n[12.2] K-means кластеризация (k=3):")
    cluster_counts = pd.Series(kmeans.labels_).value_counts().sort_index()
    for clust, count in cluster_counts.items():
        print(f"    • Кластер {clust}: {count} испытуемых")
    print("\n[12.3] Характеристика кластеров (средние значения показателей):")
    cluster_means = df.groupby('Кластер')[fa_vars].mean()
    print(np.round(cluster_means, 2).to_string())
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    dendrogram(linked, ax=axes[0], color_threshold=0.7*max(linked[:, 2]), no_labels=True)
    axes[0].set_title('Дендрограмма (метод Уорда)')
    axes[0].set_xlabel('Испытуемые')
    axes[0].set_ylabel('Евклидово расстояние')
    scatter = axes[1].scatter(scores[:, 0], scores[:, 1], c=kmeans.labels_, cmap='Set1', edgecolors='black', alpha=0.8)
    axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
    axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
    axes[1].set_title('Кластеры K-means (k=3) в пространстве PC1-PC2')
    plt.colorbar(scatter, ax=axes[1], label='Кластер')
    plt.tight_layout()

    plt.show()



# ============================================================================
# 13. ИТОГОВАЯ СВОДКА
# ============================================================================
print("\n" + "=" * 75)
print("ИТОГОВАЯ СВОДКА ПРОВЕДЕННОГО АНАЛИЗА")
print("=" * 75)

print("""
Проведенные этапы анализа:

1. ОПИСАТЕЛЬНАЯ СТАТИСТИКА
   - Рассчитаны средние, медианы, стандартные отклонения, ошибки среднего,
     коэффициенты вариации для всех количественных показателей.
   - Дана интерпретация вариабельности данных.

2. ПРОВЕРКА НОРМАЛЬНОСТИ
   - Автоматический выбор теста: Шапиро-Уилка (n ≤ 50) или
     Колмогорова-Смирнова (n > 50).
   - Q-Q графики для визуальной оценки соответствия нормальному
     распределению.

3. ГОМОГЕННОСТЬ ДИСПЕРСИЙ
   - Критерий Левене для сравнения двух групп.
   - Критерий Бартлетта для сравнения трех и более групп.
   - Автоматический выбор поправки Уэлча при нарушении гомогенности.

4. СРАВНИТЕЛЬНЫЙ АНАЛИЗ ГРУПП
   - Автоматический выбор между t-критерием Стьюдента и
     U-критерием Манна-Уитни на основе проверки нормальности.
   - Расчет размера эффекта Коэна (Cohen's d) для параметрических тестов.
   - Подробная текстовая интерпретация каждого результата.

5. ДИСПЕРСИОННЫЙ АНАЛИЗ (ANOVA)
   - Однофакторный ANOVA для факторов с ≥3 уровнями.
   - Расчет eta-squared (размер эффекта).
   - Попарные сравнения с поправкой Бонферрони.

6. АНАЛИЗ ПОВТОРНЫХ ИЗМЕРЕНИЙ
   - Автоматический выбор между парным t-критерием и
     T-критерием Вилкоксона.
   - Анализ общей выборки и отдельно по группам.

7. КОРРЕЛЯЦИОННЫЙ АНАЛИЗ
   - Матрицы корреляций Пирсона и Спирмена.
   - Интерпретация силы и направления связи.
   - Scatter plots с линией регрессии для значимых пар.

8. РЕГРЕССИОННЫЙ АНАЛИЗ
   - Простая линейная регрессия.
   - Множественная регрессия с несколькими предикторами.
   - Оценка R² и F-теста с текстовой интерпретацией.

9. АНАЛИЗ КАЧЕСТВЕННЫХ ПЕРЕМЕННЫХ
   - Таблицы сопряжености.
   - Критерий хи-квадрат Пирсона.
   - V Крамера с интерпретацией силы связи.

10. ФАКТОРНЫЙ АНАЛИЗ
    - Метод главных компонент (PCA).
    - Scree plot, биплот, факторные нагрузки.
    - Определение числа компонент для объяснения 80% дисперсии.

11. КЛАСТЕРНЫЙ АНАЛИЗ
    - Иерархическая кластеризация (дендрограмма).
    - K-means кластеризация.
    - Визуализация кластеров в PCA-пространстве.
    - Описание характеристик каждого кластера.
""")
print("=" * 75)
print("АНАЛИЗ ЗАВЕРШЕН.")
print("=" * 75)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# --- Параметры модели ---
R_max = 150   # кг — генетический потолок
R0 = 60       # кг — начальный результат
k = 0.08      # 1/неделя — скорость адаптации

# --- Дифференциальное уравнение ---
def logistic(R, t):
    return k * R * (1 - R / R_max)

# --- Численное решение ---
t = np.linspace(0, 100, 500)  # 100 недель
R_numeric = odeint(logistic, R0, t)

# --- Аналитическое решение для проверки ---
R_analytic = R_max / (1 + (R_max/R0 - 1) * np.exp(-k * t))

# --- Визуализация ---
plt.figure(figsize=(10, 6))
plt.plot(t, R_numeric, 'b-', linewidth=3, label='Численное решение (odeint)')
plt.plot(t, R_analytic, 'r--', linewidth=2, label='Аналитическое решение')
plt.axhline(R_max, color='green', linestyle=':', label=f'Потолок R_max = {R_max} кг')
plt.xlabel('Время, недели', fontsize=12)
plt.ylabel('Результат в жиме лёжа, кг', fontsize=12)
plt.title('Логистическая модель роста спортивного результата', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.show()

# --- Педагогическая интерпретация ---
print("📌 ПЕДАГОГИЧЕСКИЙ АНАЛИЗ:")
print(f"Через 10 недель: {R_analytic[50]:.1f} кг")
print(f"Через 25 недель: {R_analytic[125]:.1f} кг")
print(f"Через 50 недель: {R_analytic[250]:.1f} кг")
print(f"Через 100 недель: {R_analytic[-1]:.1f} кг (приближение к потолку)")
print("\n💡 ВЫВОД: наибольший прирост — первые 20-25 недель.")
print("   Дальше нужны принципиально новые методики (периодизация, фармподдержка).")
```

Где:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# --- Параметры модели ---
a, b, eps = 0.7, 0.8, 0.08
I = 0.5   # интенсивность нагрузки

def fitzhugh_nagumo(state, t):
    E, F = state
    dE = E - E**3/3 - F + I
    dF = eps * (E + a - b * F)
    return [dE, dF]

# --- Решение ---
t = np.linspace(0, 200, 2000)
sol = odeint(fitzhugh_nagumo, [0.1, 0.0], t)
E, F = sol[:, 0], sol[:, 1]

# --- Визуализация ---
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Динамика во времени
axes[0].plot(t, E, 'b-', linewidth=2, label='Возбуждение (работоспособность)')
axes[0].plot(t, F, 'r-', linewidth=2, label='Усталость')
axes[0].set_xlabel('Время, усл. ед.', fontsize=12)
axes[0].set_ylabel('Уровень', fontsize=12)
axes[0].set_title('Динамика работоспособности и усталости', fontsize=13)
axes[0].legend(fontsize=11)
axes[0].grid(True, alpha=0.3)

# Фазовый портрет
axes[1].plot(E, F, 'purple', linewidth=2)
axes[1].set_xlabel('Возбуждение E', fontsize=12)
axes[1].set_ylabel('Усталость F', fontsize=12)
axes[1].set_title('Фазовый портрет системы (предельный цикл)', fontsize=13)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# --- Параметры ---
k1 = 0.5   # быстрая система
k2 = 0.1   # медленная система

def energy_system(state, t):
    A, G = state
    dA = -k1 * A
    dG = k1 * A - k2 * G
    return [dA, dG]

t = np.linspace(0, 30, 500)
sol = odeint(energy_system, [100, 0], t)
A, G = sol[:, 0], sol[:, 1]

plt.figure(figsize=(11, 6))
plt.fill_between(t, 0, A, alpha=0.4, color='red', label='Анаэробная (АТФ-Кр)')
plt.fill_between(t, A, A+G, alpha=0.4, color='blue', label='Аэробная (гликолиз)')
plt.plot(t, A, 'r-', linewidth=2)
plt.plot(t, A+G, 'b-', linewidth=2, label='Общая энергия')
plt.xlabel('Время, сек', fontsize=12)
plt.ylabel('Доступная энергия, %', fontsize=12)
plt.title('Двухкомпонентная модель энергообеспечения мышц', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.show()

# --- Найти точку переключения ---
crossover = t[np.argmax(G - A)]
print(f"📌 Точка переключения анаэроб→аэроб: ~{crossover:.1f} сек")
print("💡 ВЫВОД: до этой точки спринтер работает на АТФ-Кр,")
print("   после — подключается гликолиз. Это обоснование длины отрезков в интервальной тренировке.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

def projectile(state, t, c, m, g=9.81):
    x, y, vx, vy = state
    v = np.sqrt(vx**2 + vy**2)
    dx = vx
    dy = vy
    dvx = -(c/m) * v * vx
    dvy = -g - (c/m) * v * vy
    return [dx, dy, dvx, dvy]

# --- Сравнение: вакуум vs воздух ---
t = np.linspace(0, 6, 1000)
v0, angle = 25, 45  # м/с, градусы
vx0 = v0 * np.cos(np.radians(angle))
vy0 = v0 * np.sin(np.radians(angle))

# Без сопротивления (c=0)
sol_vacuum = odeint(projectile, [0, 1.5, vx0, vy0], t, args=(0, 0.8))

# С сопротивлением (копьё — c=0.08)
sol_air = odeint(projectile, [0, 1.5, vx0, vy0], t, args=(0.08, 0.8))

plt.figure(figsize=(12, 6))
plt.plot(sol_vacuum[:, 0], sol_vacuum[:, 1], 'b--', linewidth=2, label='Вакуум (идеальная парабола)')
plt.plot(sol_air[:, 0], sol_air[:, 1], 'r-', linewidth=2, label='С сопротивлением воздуха (копьё)')
plt.axhline(0, color='black', linewidth=0.5)
plt.xlabel('Дистанция, м', fontsize=12)
plt.ylabel('Высота, м', fontsize=12)
plt.title('Траектория метания копья: влияние сопротивления воздуха', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.show()

# Анализ дальности
range_vacuum = sol_vacuum[sol_vacuum[:,1] >= 0][-1, 0]
range_air = sol_air[sol_air[:,1] >= 0][-1, 0]
print(f"📌 Дальность в вакууме: {range_vacuum:.2f} м")
print(f"📌 Дальность в воздухе: {range_air:.2f} м")
print(f"📌 Потеря из-за сопротивления: {range_vacuum - range_air:.2f} м ({(1-range_air/range_vacuum)*100:.1f}%)")
print("\n💡 ВЫВОД: оптимальный угол метания НЕ 45°, а ~42-43° — это подтверждает теория ДУ.")
```

Где:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

def lotka_volterra(state, t, alpha, beta, delta, gamma):
    X, Y = state
    dX = alpha * X - beta * X * Y
    dY = delta * X * Y - gamma * Y
    return [dX, dY]

# --- Параметры тактической модели ---
alpha, beta = 1.0, 0.1
delta, gamma = 0.075, 1.5

t = np.linspace(0, 30, 1000)
sol = odeint(lotka_volterra, [10, 5], t, args=(alpha, beta, delta, gamma))
X, Y = sol[:, 0], sol[:, 1]

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Динамика во времени
axes[0].plot(t, X, 'b-', linewidth=2, label='Атака команды А')
axes[0].plot(t, Y, 'r-', linewidth=2, label='Оборона команды Б')
axes[0].set_xlabel('Время матча, мин', fontsize=12)
axes[0].set_ylabel('Условный потенциал', fontsize=12)
axes[0].set_title('Динамика тактического преимущества', fontsize=13)
axes[0].legend(fontsize=11)
axes[0].grid(True, alpha=0.3)

# Фазовый портрет
axes[1].plot(X, Y, 'purple', linewidth=2)
axes[1].set_xlabel('Атака команды А', fontsize=12)
axes[1].set_ylabel('Оборона команды Б', fontsize=12)
axes[1].set_title('Фазовый портрет (цикличность тактики)', fontsize=13)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("💡 ВЫВОД: модель предсказывает ЦИКЛИЧНОСТЬ тактического преимущества.")
print("   Это математическое обоснование волновой тактики в футболе:")
print("   команды попеременно захватывают инициативу.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint, solve_ivp

def solve_and_plot(ode_func, y0, t_span, params=None, title="Модель"):
    """
    Универсальная функция для решения и визуализации ДУ.

    Параметры:
    -----------
    ode_func : callable — функция правой части dy/dt = f(y, t, *params)
    y0 : list — начальные условия
    t_span : tuple — (t_start, t_end)
    params : tuple — дополнительные параметры
    title : str — заголовок графика
    """
    t = np.linspace(t_span[0], t_span[1], 1000)
    if params:
        sol = odeint(ode_func, y0, t, args=params)
    else:
        sol = odeint(ode_func, y0, t)

    plt.figure(figsize=(10, 6))
    for i in range(sol.shape[1]):
        plt.plot(t, sol[:, i], linewidth=2, label=f'Переменная {i+1}')
    plt.xlabel('Время', fontsize=12)
    plt.ylabel('Значение', fontsize=12)
    plt.title(title, fontsize=14)
    plt.legend(fontsize=11)
    plt.grid(True, alpha=0.3)
    plt.show()
    return t, sol

# --- Пример применения ---
def decay_model(y, t, k):
    return -k * y

t, sol = solve_and_plot(
    decay_model,
    y0=[100],
    t_span=(0, 20),
    params=(0.15,),
    title="Модель восстановления после нагрузки"
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

```python
def моделировать(формула, старт, конец, параметры=None,
                 название="Динамика показателя",
                 подписи_осей=("Время", "Значение"),
                 сравнить_с=None):
    """
    УНИВЕРСАЛЬНЫЙ КОНСТРУКТОР МОДЕЛЕЙ СПОРТИВНОЙ ДИНАМИКИ

    Параметры:
    ----------
    формула : функция
        Формула скорости изменения. Пишется так:
        def формула(значение, время, параметры):
            return 0.5 * значение - 0.01 * значение**2

    старт : число
        Начальное значение (например, стартовый результат спортсмена)

    конец : число
        До какого времени считать (например, до 40 недель)

    параметры : tuple или None
        Дополнительные константы для формулы (необязательно)

    название : str
        Заголовок графика

    подписи_осей : tuple
        (что по X, что по Y) — например ("Недели", "Результат, кг")

    сравнить_с : dict или None
        Для сравнения моделей: {"метка": значение_старта}

    Возвращает:
    -----------
    t : массив времени
    решение : массив значений
    """

    # ---- Шаг 1: Создаём массив времени ----
    t = np.linspace(0, конец, 500)

    # ---- Шаг 2: Решаем уравнение ----
    if параметры:
        # Если есть доп. параметры, передаём их в формулу
        решение = odeint(формула, старт, t, args=параметры)
    else:
        # Если нет — решаем как есть
        решение = odeint(формула, старт, t)

    # ---- Шаг 3: Достаём значения (первая колонка) ----
    значения = решение[:, 0]

    # ---- Шаг 4: Рисуем график ----
    plt.figure(figsize=(11, 6))
    plt.plot(t, значения, 'b-', linewidth=3, label='Основной прогноз')

    # Если нужно сравнить с другой моделью
    if сравнить_с:
        for метка, другой_старт in сравнить_с.items():
            if параметры:
                другое_решение = odeint(формула, другой_старт, t, args=параметры)
            else:
                другое_решение = odeint(формула, другой_старт, t)
            plt.plot(t, другое_решение[:, 0], '--', linewidth=2, label=метка)

    plt.xlabel(подписи_осей[0], fontsize=13)
    plt.ylabel(подписи_осей[1], fontsize=13)
    plt.title(название, fontsize=15)
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=12)
    plt.show()

    # ---- Шаг 5: Выводим итог ----
    print(f"\n📊 РЕЗУЛЬТАТ МОДЕЛИРОВАНИЯ:")
    print(f"   Стартовое значение:    {значения[0]:.2f}")
    print(f"   Середина ({конец//2} ед.):    {значения[len(значения)//2]:.2f}")
    print(f"   Финал ({конец} ед.):       {значения[-1]:.2f}")
    print(f"   Изменение:             {значения[-1] - значения[0]:+.2f}")

    return t, значения
```

```python
# ============================================================
# УМНЫЙ КОНСТРУКТОР ДИФФЕРЕНЦИАЛЬНЫХ МОДЕЛЕЙ В СПОРТЕ
# С автоматическим выводом уравнений, графиков и интерпретаций
# ============================================================

!pip install ipython -q

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
from IPython.display import display, Markdown, Latex, HTML

# ============================================================
# БАЗА ЗНАНИЙ: ОПИСАНИЯ ВСЕХ МОДЕЛЕЙ
# ============================================================

БАЗА_МОДЕЛЕЙ = {
    "логистический_рост": {
        "название": "Логистический рост спортивного результата",
        "ученый": "Пьер Франсуа Ферхюльст (1838)",
        "формула_latex": r"\frac{dR}{dt} = k \cdot R \cdot \left(1 - \frac{R}{R_{max}}\right)",
        "решение_latex": r"R(t) = \frac{R_{max}}{1 + \left(\frac{R_{max}}{R_0} - 1\right) e^{-kt}}",
        "символы": {
            "R(t)": "текущий результат спортсмена",
            "t": "время тренировок (обычно в неделях)",
            "k": "скорость адаптации к нагрузке (чем больше — тем быстрее рост)",
            "R_max": "генетический потолок (максимально достижимый результат)",
            "R_0": "начальный результат спортсмена"
        },
        "смысл": "Рост сначала идёт быстро (новичок), затем замедляется по мере приближения к потолку (элита).",
        "применение": "Прогноз роста силовых показателей, выносливости, скорости."
    },

    "восстановление": {
        "название": "Модель восстановления после нагрузки / травмы",
        "ученый": "Классическая модель экспоненциального восстановления",
        "формула_latex": r"\frac{dR}{dt} = -k \cdot (R - R_{покой})",
        "решение_latex": r"R(t) = R_{покой} + (R_0 - R_{покой}) e^{-kt}",
        "символы": {
            "R(t)": "текущее состояние спортсмена (сила, пульс, работоспособность)",
            "t": "время восстановления",
            "k": "скорость восстановления (чем больше — тем быстрее возврат к норме)",
            "R_покой": "уровень покоя (нормальный уровень)",
            "R_0": "состояние сразу после нагрузки/травмы"
        },
        "смысл": "Спортсмен экспоненциально возвращается к своему нормальному состоянию. Чем дальше от нормы — тем быстрее идёт восстановление.",
        "применение": "Реабилитация после травм, восстановление пульса, расчёт интервалов отдыха."
    },

    "обучение_забывание": {
        "название": "Модель обучения с забыванием навыка",
        "ученый": "Модель Эббингауза (адаптированная)",
        "формула_latex": r"\frac{dR}{dt} = A - \lambda \cdot R",
        "решение_latex": r"R(t) = \frac{A}{\lambda} + \left(R_0 - \frac{A}{\lambda}\right) e^{-\lambda t}",
        "символы": {
            "R(t)": "уровень освоенного навыка",
            "t": "время",
            "A": "скорость обучения (сколько добавляется за единицу времени)",
            "λ (лямбда)": "коэффициент забывания (доля потерянного навыка за единицу времени)",
            "R_0": "начальный уровень навыка"
        },
        "смысл": "Навык растёт от тренировок, но параллельно частично забывается. Равновесие наступает, когда обучение = забывание.",
        "применение": "Планирование перерывов в тренировках, оценка потери формы после отпуска."
    },

    "перетренированность": {
        "название": "Модель перетренированности (усталость съедает рост)",
        "ученый": "Нелинейная модель с квадратичным торможением",
        "формула_latex": r"\frac{dR}{dt} = A - \beta \cdot R^2",
        "решение_latex": r"R(t) = \sqrt{\frac{A}{\beta}} \cdot \tanh\left(\sqrt{A\beta} \cdot t + C\right)",
        "символы": {
            "R(t)": "текущий результат",
            "t": "время",
            "A": "прирост от тренировок (положительный вклад)",
            "β (бета)": "коэффициент усталости (чем больше — тем сильнее тормозит)",
            "R^2": "квадратичный член — моделирует резкое нарастание усталости"
        },
        "смысл": "Когда результат становится слишком высоким, усталость растёт быстрее, чем прирост. Наступает плато или падение.",
        "применение": "Диагностика перетренированности, обоснование периодов разгрузки."
    },

    "двухкомпонентная_энергия": {
        "название": "Двухкомпонентная модель энергообеспечения мышц",
        "ученый": "Модель АТФ-Кр / гликолиз (Маргария–Эдвардс)",
        "формула_latex": r"\begin{cases} \frac{dA}{dt} = -k_1 A \\ \frac{dG}{dt} = k_1 A - k_2 G \end{cases}",
        "решение_latex": r"A(t) = A_0 e^{-k_1 t}, \quad G(t) = \frac{k_1 A_0}{k_2 - k_1}(e^{-k_1 t} - e^{-k_2 t})",
        "символы": {
            "A(t)": "запас быстрой энергии (АТФ-креатинфосфат)",
            "G(t)": "запас медленной энергии (гликолиз)",
            "k_1": "скорость расходования быстрой энергии",
            "k_2": "скорость расходования медленной энергии",
            "A_0": "начальный запас АТФ-Кр"
        },
        "смысл": "Два источника энергии работают последовательно: сначала быстро опустошается АТФ-Кр, затем подключается гликолиз.",
        "применение": "Обоснование длины отрезков в интервальной тренировке (спринт, плавание)."
    },

    "баллистика": {
        "название": "Баллистика снаряда с сопротивлением воздуха",
        "ученый": "Модель Ньютона с квадратичным сопротивлением",
        "формула_latex": r"\begin{cases} \frac{dv_x}{dt} = -\frac{c}{m} v \cdot v_x \\ \frac{dv_y}{dt} = -g - \frac{c}{m} v \cdot v_y \end{cases}",
        "решение_latex": r"\text{Решается численно (аналитического решения нет)}",
        "символы": {
            "v_x, v_y": "проекции скорости снаряда",
            "v": "модуль скорости: √(v_x² + v_y²)",
            "c": "коэффициент сопротивления воздуха (зависит от формы снаряда)",
            "m": "масса снаряда",
            "g": "ускорение свободного падения = 9.81 м/с²"
        },
        "смысл": "Снаряд летит не по идеальной параболе — воздух тормозит его. Оптимальный угол вылета — не 45°, а меньше.",
        "применение": "Метание копья, диска, толкание ядра, удар по мячу."
    },

    "хищник_жертва": {
        "название": "Модель тактического противостояния (Лотка–Вольтерра)",
        "ученый": "Альфред Лотка и Вито Вольтерра (1925–1926)",
        "формула_latex": r"\begin{cases} \frac{dX}{dt} = \alpha X - \beta X Y \\ \frac{dY}{dt} = \delta X Y - \gamma Y \end{cases}",
        "решение_latex": r"\text{Циклическое решение (предельный цикл)}",
        "символы": {
            "X(t)": "атакующий потенциал команды А",
            "Y(t)": "оборонительный потенциал команды Б",
            "α (альфа)": "скорость наращивания атаки",
            "β (бета)": "эффективность обороны",
            "δ (дельта)": "конверсия обороны в контратаку",
            "γ (гамма)": "усталость обороняющихся"
        },
        "смысл": "Тактическое преимущество в игре циклично: за атакой всегда следует контратака. Это математическая основа волновой тактики.",
        "применение": "Тактическое планирование в футболе, баскетболе, хоккее."
    }
}


# ============================================================
# ФУНКЦИЯ КРАСИВОГО ВЫВОДА УРАВНЕНИЯ
# ============================================================

def показать_модель(ключ_модели):
    """Печатает название, уравнение, расшифровку символов."""
    инфо = БАЗА_МОДЕЛЕЙ[ключ_модели]

    # Заголовок
    display(Markdown(f"### 📐 {инфо['название']}"))
    display(Markdown(f"*Автор модели: {инфо['ученый']}*"))

    # Само уравнение
    display(Markdown("**🧮 Дифференциальное уравнение:**"))
    display(Latex(f"$${инфо['формула_latex']}$$"))

    # Аналитическое решение (если есть)
    if инфо['решение_latex']:
        display(Markdown("**✅ Аналитическое решение (если существует):**"))
        display(Latex(f"$${инфо['решение_latex']}$$"))

    # Расшифровка символов
    display(Markdown("**📖 Расшифровка символов:**"))
    символы_md = "\n".join([f"- `${k}` — {v}" for k, v in инфо['символы'].items()])
    display(Markdown(символы_md))

    # Смысл и применение
    display(Markdown(f"**💡 Физический смысл:** {инфо['смысл']}"))
    display(Markdown(f"**🎯 Применение в спорте:** {инфо['применение']}"))
    display(Markdown("---"))


# ============================================================
# УНИВЕРСАЛЬНЫЙ КОНСТРУКТОР (С ВЫВОДОМ УРАВНЕНИЯ)
# ============================================================

def моделировать(ключ_модели, формула, старт, конец, параметры=None,
                 подписи_осей=("Время", "Значение"),
                 сравнить_с=None):
    """
    УМНЫЙ КОНСТРУКТОР: выводит уравнение, график и результаты.

    Параметры:
    ----------
    ключ_модели : str — ключ из БАЗА_МОДЕЛЕЙ (например, 'логистический_рост')
    формула : callable — функция правой части ДУ
    старт : число — начальное значение
    конец : число — до какого времени считать
    параметры : tuple — дополнительные параметры формулы
    подписи_осей : tuple — (что по X, что по Y)
    сравнить_с : dict — {"метка": старт_значение} для сравнения
    """

    # ============ ШАГ 1: ПОКАЗЫВАЕМ УРАВНЕНИЕ ============
    показать_модель(ключ_модели)

    # ============ ШАГ 2: РЕШАЕМ УРАВНЕНИЕ ============
    t = np.linspace(0, конец, 500)

    if параметры:
        решение = odeint(формула, старт, t, args=параметры)
    else:
        решение = odeint(формула, старт, t)

    значения = решение[:, 0]

    # ============ ШАГ 3: РИСУЕМ ГРАФИК ============
    plt.figure(figsize=(11, 6))
    plt.plot(t, значения, 'b-', linewidth=3, label='Основной прогноз')

    if сравнить_с:
        цвета = ['g', 'r', 'orange', 'purple', 'brown']
        for i, (метка, другой_старт) in enumerate(сравнить_с.items()):
            if параметры:
                др = odeint(формула, другой_старт, t, args=параметры)
            else:
                др = odeint(формула, другой_старт, t)
            plt.plot(t, др[:, 0], '--', color=цвета[i % len(цвета)],
                     linewidth=2, label=метка)

    plt.xlabel(подписи_осей[0], fontsize=13)
    plt.ylabel(подписи_осей[1], fontsize=13)
    plt.title(БАЗА_МОДЕЛЕЙ[ключ_модели]['название'], fontsize=14)
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=12)
    plt.show()

    # ============ ШАГ 4: ЧИСЛОВЫЕ РЕЗУЛЬТАТЫ ============
    display(Markdown("**📊 Численные результаты:**"))
    результат_md = f"""
| Точка | Время | Значение | Изменение от старта |
|:------|:------|:---------|:--------------------|
| Старт | 0 | {значения[0]:.2f} | — |
| 25% | {конец*0.25:.1f} | {значения[int(len(значения)*0.25)]:.2f} | {значения[int(len(значения)*0.25)] - значения[0]:+.2f} |
| Середина | {конец/2:.1f} | {значения[len(значения)//2]:.2f} | {значения[len(значения)//2] - значения[0]:+.2f} |
| 75% | {конец*0.75:.1f} | {значения[int(len(значения)*0.75)]:.2f} | {значения[int(len(значения)*0.75)] - значения[0]:+.2f} |
| Финал | {конец} | {значения[-1]:.2f} | {значения[-1] - значения[0]:+.2f} |
"""
    display(Markdown(результат_md))

    return t, значения


# ============================================================
# ПРИМЕР 1: ЛОГИСТИЧЕСКИЙ РОСТ В ЖИМЕ ЛЁЖА
# ============================================================

def рост_жима(R, t):
    k = 0.08
    Потолок = 150
    return k * R * (1 - R / Потолок)

t, R = моделировать(
    ключ_модели = "логистический_рост",
    формула = рост_жима,
    старт = 60,
    конец = 40,
    подписи_осей = ("Недели тренировок", "Результат в жиме лёжа, кг"),
    сравнить_с = {
        "Новичок (60 кг)": 60,
        "Средний (90 кг)": 90,
        "Элита (120 кг)": 120
    }
)
```

## Восстановление после травмы (с полным выводом)

```python
def восстановление(R, t):
    k = 0.3
    Норма = 100
    return -k * (R - Норма)

t, R = моделировать(
    ключ_модели = "восстановление",
    формула = восстановление,
    старт = 30,
    конец = 20,
    подписи_осей = ("Недели реабилитации", "Сила, % от нормы"),
    сравнить_с = {
        "Тяжёлая травма (20%)": 20,
        "Средняя травма (40%)": 40,
        "Лёгкая травма (70%)": 70
    }
)
```

## Перетренированность (нелинейная модель)

```python
def перетренированность(R, t):
    A = 0.3
    beta = 0.005
    return A - beta * R**2

t, R = моделировать(
    ключ_модели = "перетренированность",
    формула = перетренированность,
    старт = 2,
    конец = 30,
    подписи_осей = ("Недели", "Уровень результата"),
    сравнить_с = {
        "Мягкий режим (β=0.002)": 2,
    }
)
```

Обратите внимание: формула β внутри функции не поменяется просто из-за метки в сравнить_с. Если хотите реально сравнить с другим β — нужно передать его через параметры:

```python
def перетренированность(R, t, A, beta):
    return A - beta * R**2

# Одна модель с beta=0.005
t, R = моделировать(
    ключ_модели = "перетренированность",
    формула = перетренированность,
    старт = 2,
    конец = 30,
    параметры = (0.3, 0.005),
    подписи_осей = ("Недели", "Уровень")
)
```

## Двухкомпонентная энергия (система уравнений)

```python
def моделировать_систему(ключ_модели, формула, старт_список, конец,
                          параметры=None, подписи_переменных=None):
    """Для систем ДУ (например, АТФ-Кр + гликолиз)."""
    показать_модель(ключ_модели)

    t = np.linspace(0, конец, 500)
    решение = odeint(формула, старт_список, t, args=параметры or ())

    if подписи_переменных is None:
        подписи_переменных = [f'Переменная {i+1}' for i in range(решение.shape[1])]

    plt.figure(figsize=(11, 6))
    цвета = ['red', 'blue', 'green', 'orange', 'purple']
    for i, метка in enumerate(подписи_переменных):
        plt.plot(t, решение[:, i], color=цвета[i % len(цвета)],
                 linewidth=3, label=метка)
    plt.xlabel('Время', fontsize=13)
    plt.ylabel('Значение', fontsize=13)
    plt.title(БАЗА_МОДЕЛЕЙ[ключ_модели]['название'], fontsize=14)
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.show()

    return t, решение


# --- Пример: АТФ-Кр + гликолиз ---
def энергия(state, t, k1, k2):
    A, G = state
    dA = -k1 * A
    dG = k1 * A - k2 * G
    return [dA, dG]

t, sol = моделировать_систему(
    ключ_модели = "двухкомпонентная_энергия",
    формула = энергия,
    старт_список = [100, 0],
    конец = 30,
    параметры = (0.5, 0.1),
    подписи_переменных = ["АТФ-Кр (быстрая энергия)", "Гликолиз (медленная энергия)"]
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Настройка стиля
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

# --- Сценарий ---
# Мы проверяем, отличается ли средний рост группы от гипотетического μ0 = 170 см
# H0: средний рост = 170 см
# H1: средний рост ≠ 170 см (двусторонний тест)

mu0 = 170          # среднее при нулевой гипотезе
sample_mean = 173.5  # наблюдаемое среднее в выборке
sample_std = 6     # стандартное отклонение в выборке
n = 30             # размер выборки

# Расчет t-статистики
se = sample_std / np.sqrt(n)
t_obs = (sample_mean - mu0) / se

# Степени сободы
df = n - 1

# p-value для двустороннего теста
p_value = 2 * (1 - stats.t.cdf(abs(t_obs), df))

print(f"Наблюдаемое среднее: {sample_mean} см")
print(f"Стандартная ошибка: {se:.3f}")
print(f"t-статистика: {t_obs:.3f}")
print(f"p-value: {p_value:.4f}")

# --- Построение графика ---
fig, ax = plt.subplots(figsize=(12, 7))

# Диапазон значений для t-распределения
x = np.linspace(-4.5, 4.5, 1000)
y = stats.t.pdf(x, df)

# Рисуем нулевое распределение (t-распределение при H0)
ax.plot(x, y, 'b-', linewidth=2, label='Нулевое распределение (t при H0)')
ax.fill_between(x, 0, y, alpha=0.1, color='blue')

# Вертикальная линия наблюдаемого t
ax.axvline(t_obs, color='red', linestyle='--', linewidth=2, label=f'Наблюдаемое t = {t_obs:.2f}')
ax.axvline(-t_obs, color='red', linestyle='--', linewidth=2)

# Заштриховываем области, соответствующие p-value (два хвоста)
# Правый хвост
x_right = x[x >= t_obs]
y_right = stats.t.pdf(x_right, df)
ax.fill_between(x_right, 0, y_right, color='green', alpha=0.5, label=f'p-value область = {p_value:.4f}')

# Левый хвост (симметричный)
x_left = x[x <= -t_obs]
y_left = stats.t.pdf(x_left, df)
ax.fill_between(x_left, 0, y_left, color='green', alpha=0.5)

# Добавляем аннотации
ax.annotate(f'Наблюдаемое\nзначение\nt = {t_obs:.2f}',
            xy=(t_obs, stats.t.pdf(t_obs, df)),
            xytext=(t_obs + 1, stats.t.pdf(t_obs, df) + 0.15),
            arrowprops=dict(arrowstyle='->', color='red'),
            fontsize=11, color='red')

ax.annotate('p-value = суммарная площадь\nзеленых хвостов',
            xy=(2.8, 0.03), fontsize=12, color='darkgreen',
            bbox=dict(boxstyle='round,pad=0.5', facecolor='lightgreen', alpha=0.8))

# Добавляем легенду уровня начимости
alpha = 0.05
t_crit = stats.t.ppf(1 - alpha/2, df)
ax.axvline(t_crit, color='orange', linestyle=':', linewidth=1.5, alpha=0.8)
ax.axvline(-t_crit, color='orange', linestyle=':', linewidth=1.5, alpha=0.8)
ax.annotate(f'Критическое значение\n(α = 0.05)', xy=(t_crit, 0.05), color='orange', fontsize=10)

# Оформление
ax.set_title('Наглядная демонстрация p-value', fontsize=16, pad=20)
ax.set_xlabel('t-статистика', fontsize=12)
ax.set_ylabel('Плотность вероятности', fontsize=12)
ax.legend(loc='upper right', fontsize=10)
ax.set_xlim(-4.5, 4.5)
ax.set_ylim(0, max(y) * 1.25)
ax.grid(True, alpha=0.3)

# Добавляем текстовое объяснение
explanation = (
    "p-value — это вероятность получить такое же или более экстремальное значение\n"
    "статистики, как наблюдаемое, если нулевая гипотеза (H₀) верна.\n\n"
    f"В данном примере: если истинное среднее = {mu0} см,\n"
    f"то вероятность получить среднее ≥ {sample_mean} или ≤ {mu0 - (sample_mean - mu0)} см\n"
    f"составляет p = {p_value:.4f}."
)
fig.text(0.5, -0.15, explanation, ha='center', fontsize=11,
         bbox=dict(boxstyle='round,pad=0.7', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.savefig('p_value_demo.png', dpi=150, bbox_inches='tight', facecolor='white')
plt.show()
print("График сохранен!")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)

# --- Генерация данных для двух групп ---
# Контрольная группа
n_control = 100
mean_control = 170  # см (например, рост)
std_control = 10
control = np.random.normal(mean_control, std_control, n_control)

# Экспериментальная группа (чуть выше)
n_exp = 100
mean_exp = 174  # см
std_exp = 10
experimental = np.random.normal(mean_exp, std_exp, n_exp)

# --- Расчет статистик ---
x1_bar = np.mean(control)
x2_bar = np.mean(experimental)
s1 = np.std(control, ddof=1)
s2 = np.std(experimental, ddof=1)

# Стандартная ошибка разности
se_diff = np.sqrt(s1**2 / n_control + s2**2 / n_exp)

# t-статистика (Welch's t-test)
t_obs = (x2_bar - x1_bar) / se_diff

# Степени свободы (Welch–Satterthwaite)
df = (s1**2 / n_control + s2**2 / n_exp)**2 / (
    (s1**2 / n_control)**2 / (n_control - 1) + (s2**2 / n_exp)**2 / (n_exp - 1)
)

# p-value (двусторонний)
p_value = 2 * (1 - stats.t.cdf(abs(t_obs), df))

print("=" * 50)
print("СРАВНЕНИ ДВУХ ГРУПП (t-тест)")
print("=" * 50)
print(f"Контрольная группа:  n={n_control}, среднее={x1_bar:.2f}, std={s1:.2f}")
print(f"Экспериментальная:   n={n_exp}, среднее={x2_bar:.2f}, std={s2:.2f}")
print(f"Разность средних:    {x2_bar - x1_bar:.2f}")
print(f"Стандартная ошибка:  {se_diff:.3f}")
print(f"t-статистика:        {t_obs:.3f}")
print(f"Степени свободы:     {df:.1f}")
print(f"p-value:             {p_value:.4f}")
print("=" * 50)

# --- Визуализация ---
fig, axes = plt.subplots(2, 2, figsize=(22, 10), gridspec_kw={'width_ratios': [2, 1]})
fig.suptitle('p-value при сравнении двух групп (контрольная vs экспериментальная)',
             fontsize=16, fontweight='bold', y=0.98)

# ===== График 1: Распределение данных =====
ax1 = axes[0, 0]
bins = np.linspace(140, 210, 35)
ax1.hist(control, bins=bins, alpha=0.6, color='steelblue', label='Контрольная', edgecolor='white')
ax1.hist(experimental, bins=bins, alpha=0.6, color='coral', label='Экспериментальная', edgecolor='white')
ax1.axvline(x1_bar, color='steelblue', linestyle='--', linewidth=2)
ax1.axvline(x2_bar, color='coral', linestyle='--', linewidth=2)
ax1.annotate(f'Среднее\n{x1_bar:.1f}', xy=(x1_bar, ax1.get_ylim()[1]*0.8),
             ha='center', color='steelblue', fontweight='bold')
ax1.annotate(f'Среднее\n{x2_bar:.1f}', xy=(x2_bar, ax1.get_ylim()[1]*0.8),
             ha='center', color='coral', fontweight='bold')
ax1.set_title('1. Распределение данных в группах')
ax1.set_xlabel('Значение')
ax1.set_ylabel('Частота')
ax1.legend()
ax1.grid(True, alpha=0.3)

# ===== График 2: Доверительные интервалы средних =====
ax2 = axes[0, 1]
ci1 = stats.t.interval(0.95, n_control-1, loc=x1_bar, scale=s1/np.sqrt(n_control))
ci2 = stats.t.interval(0.95, n_exp-1, loc=x2_bar, scale=s2/np.sqrt(n_exp))

y_pos = [1, 2]
means = [x1_bar, x2_bar]
errors = [[x1_bar - ci1[0], x2_bar - ci2[0]], [ci1[1] - x1_bar, ci2[1] - x2_bar]]
colors = ['steelblue', 'coral']

ax2.barh(y_pos, means, xerr=errors, color=colors, alpha=0.7, capsize=8, edgecolor='black')
ax2.set_yticks(y_pos)
ax2.set_yticklabels(['Контрольная', 'Экспериментальная'])
ax2.set_xlabel('Среднее значение')
ax2.set_title('2. Средние с 95% доверительными интервалами')
ax2.grid(True, alpha=0.3, axis='x')

# Подписи значений
for i, (m, c) in enumerate(zip(means, colors)):
    ax2.text(m, i+1.25, f'{m:.2f}', ha='center', color=c, fontweight='bold')

# ===== График 3: Нулевое распределение t-статистики =====
ax3 = axes[1, 0]
x_t = np.linspace(-4.5, 4.5, 1000)
y_t = stats.t.pdf(x_t, df)

ax3.plot(x_t, y_t, 'b-', linewidth=2, label='Нулевое распределение t (при H₀: разность = 0)')
ax3.fill_between(x_t, 0, y_t, alpha=0.1, color='blue')

# Области p-value
x_right = x_t[x_t >= t_obs]
y_right = stats.t.pdf(x_right, df)
ax3.fill_between(x_right, 0, y_right, color='green', alpha=0.5, label=f'p-value = {p_value:.4f}')

x_left = x_t[x_t <= -t_obs]
y_left = stats.t.pdf(x_left, df)
ax3.fill_between(x_left, 0, y_left, color='green', alpha=0.5)

ax3.axvline(t_obs, color='red', linestyle='--', linewidth=2)
ax3.axvline(-t_obs, color='red', linestyle='--', linewidth=2)
ax3.annotate(f'Наблюдаемое\nt = {t_obs:.2f}', xy=(t_obs, stats.t.pdf(t_obs, df)),
             xytext=(t_obs+0.5, stats.t.pdf(t_obs, df)+0.1),
             arrowprops=dict(arrowstyle='->', color='red'), color='red', fontweight='bold')

ax3.set_title('3. Нулевое распределение t-статистики')
ax3.set_xlabel('t')
ax3.set_ylabel('Плотность вероятности')
ax3.legend(loc='upper right')
ax3.grid(True, alpha=0.3)
ax3.set_xlim(-4.5, 4.5)

# ===== График 4: Наглядная схема p-value =====
ax4 = axes[1, 1]
ax4.set_xlim(0, 1)
ax4.set_ylim(0, 1)
ax4.axis('off')

# Текстовое объяснение
textstr = (
    "Что такое p-value?\n"
    "────────────────────────────\n\n"
    "H₀ (нулевая гипотеза):\n\n"
    "  Средние групп НЕ отличаются\n"
    "  (разность = 0)\n\n"
    "H₁ (альтернативная):\n"
    "  Средние ОТЛИЧАЮТСЯ\n"
    "  (разность ≠ 0)\n\n"
    f"Наблюдаемая разность: {x2_bar - x1_bar:.2f}\n\n"
    f"p-value = {p_value:.4f}\n\n"
    "Интерпретация:\n"
)

if p_value < 0.05:
    textstr += (
        f"  p < 0.05 → отвергам H₀\n"
        f"  Разница СТАТИСТИЧЕСКИ\n"
        f"  ЗНАЧИМА!"
    )
else:
    textstr += (
        f"  p ≥ 0.05 → НЕ отвергаем H₀\n"
        f"  Разница НЕ значима."
    )

textstr += (
    "\n\n"
    "p-value = вероятность получить\n"
    "ТАКУЮ или БОЛЕЕ сильную разницу\n"
    "между группами, если на самом деле\n"
    "разницы НЕТ (H₀ верна)."
)

props = dict(boxstyle='round,pad=0.8', facecolor='wheat', alpha=0.9)
ax4.text(0.5, 0.5, textstr, transform=ax4.transAxes, fontsize=12,
         verticalalignment='center', horizontalalignment='center', bbox=props)
ax4.set_title('4. Интерпретация результата')

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.savefig('p_value_two_groups.png', dpi=150, bbox_inches='tight', facecolor='white')
plt.show()
print("График сохранен!")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)

# Создаем три разных распределения для наглядности
n = 200

# 1. Нормальное распределение (симметричное)
normal = np.random.normal(100, 10, n)

# 2. Смещенное вправо (right-skewed) — лог-нормальное
skewed = np.random.lognormal(4.5, 0.5, n)

# 3. С тяжелыми хвостами — смесь нормальных
heavy_tailed = np.concatenate([
    np.random.normal(100, 5, int(n*0.85)),
    np.random.normal(100, 40, int(n*0.15))
])

datasets = {
    'Нормальное (симметричное)': normal,
    'Смещенное вправо': skewed,
    'С тяжелыми хвостами': heavy_tailed
}

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Распределения, доверительные интервалы и Box Plot',
             fontsize=16, fontweight='bold', y=0.98)

colors = ['steelblue', 'coral', 'mediumseagreen']

for i, (name, data) in enumerate(datasets.items()):
    color = colors[i]

    # ===== ВЕРХНИЙ РЯД: Гистограмма + KDE + доверительный интервал =====
    ax_top = axes[0, i]

    # Гистограмма
    counts, bins, patches = ax_top.hist(data, bins=35, density=True,
                                         alpha=0.5, color=color, edgecolor='white')

    # KDE (кривая плотности)
    from scipy.stats import gaussian_kde
    kde = gaussian_kde(data)
    x_range = np.linspace(data.min() - 5, data.max() + 5, 500)
    ax_top.plot(x_range, kde(x_range), color='darkblue', linewidth=2, label='KDE')

    # Среднее и медиана
    mean_val = np.mean(data)
    median_val = np.median(data)
    ax_top.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Среднее = {mean_val:.1f}')
    ax_top.axvline(median_val, color='green', linestyle='-.', linewidth=2, label=f'Медиана = {median_val:.1f}')

    # 95% доверительный интервал для среднего
    se = stats.sem(data)  # стандартная ошибка среднего
    ci = stats.t.interval(0.95, len(data)-1, loc=mean_val, scale=se)
    ax_top.axvspan(ci[0], ci[1], alpha=0.2, color='red', label=f'95% ДИ: [{ci[0]:.1f}, {ci[1]:.1f}]')

    # Аннотация
    ax_top.annotate(f'95% ДИ\nсреднего', xy=((ci[0]+ci[1])/2, ax_top.get_ylim()[1]*0.9),
                    ha='center', fontsize=9, color='darkred', fontweight='bold')

    ax_top.set_title(f'{name}\n(n={len(data)})', fontsize=12, fontweight='bold')
    ax_top.set_xlabel('Значение')
    ax_top.set_ylabel('Плотность')
    ax_top.legend(loc='upper right', fontsize=8)
    ax_top.grid(True, alpha=0.3)

    # ===== НИЖНИЙ РЯД: Box Plot =====
    ax_bot = axes[1, i]

    bp = ax_bot.boxplot(data, vert=False, patch_artist=True, widths=0.5,
                        boxprops=dict(facecolor=color, alpha=0.7, color='black'),
                        medianprops=dict(color='green', linewidth=2),
                        whiskerprops=dict(color='black', linewidth=1.5),
                        capprops=dict(color='black', linewidth=1.5),
                        flierprops=dict(marker='o', markerfacecolor='red', markersize=4, alpha=0.5))

    # Среднее на box plot
    ax_bot.plot(mean_val, 1, 'D', color='red', markersize=10, label=f'Среднее = {mean_val:.1f}')
    ax_bot.axvline(mean_val, color='red', linestyle='--', alpha=0.5)

    # Доверительный интервал на box plot
    ax_bot.axvspan(ci[0], ci[1], alpha=0.15, color='red')
    ax_bot.annotate('95% ДИ\nсреднего', xy=((ci[0]+ci[1])/2, 1.35),
                    ha='center', fontsize=9, color='darkred', fontweight='bold')

    # Подписи квантилей
    q1, q2, q3 = np.percentile(data, [25, 50, 75])
    ax_bot.annotate(f'Q1={q1:.1f}', xy=(q1, 0.65), ha='center', fontsize=9, color='darkgreen')
    ax_bot.annotate(f'Медиана={q2:.1f}', xy=(q2, 0.55), ha='center', fontsize=9, color='green', fontweight='bold')
    ax_bot.annotate(f'Q3={q3:.1f}', xy=(q3, 0.65), ha='center', fontsize=9, color='darkgreen')

    ax_bot.set_title(f'Box Plot: {name}', fontsize=12, fontweight='bold')
    ax_bot.set_xlabel('Значение')
    ax_bot.set_yticks([1])
    ax_bot.set_yticklabels([''])
    ax_bot.legend(loc='lower right', fontsize=9)
    ax_bot.grid(True, alpha=0.3, axis='x')

    # Статистики в консоль
    print(f"\n{name}:")
    print(f"  Среднее = {mean_val:.2f}, Медиана = {median_val:.2f}")
    print(f"  Станд. ошибка среднего = {se:.3f}")
    print(f"  95% ДИ для среднего = [{ci[0]:.2f}, {ci[1]:.2f}]")
    print(f"  Станд. отклонение = {np.std(data, ddof=1):.2f}")

# Добавляем текстовое объяснение внизу
explanation = (
    "Что показывает эта визуализация:\n\n"
    "• ВЕРХНИЙ ряд — форма распределения данных (гистограмма + KDE).\n"
    "  Красная пунктир = среднее, зеленая штрих-пунктир = медиана.\n"
    "  Красная полоса = 95% доверительный интервал (ДИ) для среднего —\n"
    "  диапазон, в котором с 95% уверенностью лежит ИСТИННОЕ среднее генеральной совокупности.\n\n"
    "• НИЖНИЙ ряд — Box Plot (ящик с усами):\n"
    "  ▬ Зеленая линия в ящике = медиана (50-й процентиль)\n"
    "  ▬ Границы ящика = Q1 (25%) и Q3 (75%), IQR = Q3 - Q1\n"
    "  ▬ Усы = Q1 - 1.5·IQR и Q3 + 1.5·IQR\n"
    "  ▬ Красные точки = выбросы за пределами усов\n"
    "  ▬ Красный ромб = среднее, красная полоса = 95% ДИ среднего\n\n"
    "КЛЮЧЕВОЙ ВЫВОД: ДИ среднего показывает точность оценки среднего.\n"
    "Чем уже ДИ, тем точнее наша оценка. ДИ зависит от объема выборки и разброса данных."
)

fig.text(0.5, -0.22, explanation, ha='center', fontsize=10,
         bbox=dict(boxstyle='round,pad=0.8', facecolor='lightyellow', alpha=0.9))

plt.tight_layout(rect=[0, 0.02, 1, 0.94])
plt.savefig('distributions_ci_boxplot.png', dpi=150, bbox_inches='tight', facecolor='white')
plt.show()
print("\nГрафик сохранен!")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)

# Истинные параметры генеральной совокупности
TRUE_MEAN = 100
TRUE_STD = 15

# Три выборки разного размера
sample_sizes = [30, 100, 500]
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Как размер выборки влияет на доверительный интервал\n(сверху — распределение, снизу — Box Plot)',
             fontsize=15, fontweight='bold', y=0.98)

colors = ['coral', 'steelblue', 'mediumseagreen']

for i, n in enumerate(sample_sizes):
    sample = np.random.normal(TRUE_MEAN, TRUE_STD, n)
    mean_val = np.mean(sample)
    std_val = np.std(sample, ddof=1)
    se = std_val / np.sqrt(n)
    ci_lower, ci_upper = stats.t.interval(0.95, n-1, loc=mean_val, scale=se)

    # ===== ВЕРХ: Гистограмма + KDE + ДИ =====
    ax_top = axes[0, i]

    bins = np.linspace(min(sample.min(), TRUE_MEAN - 50), max(sample.max(), TRUE_MEAN + 50), 30)
    ax_top.hist(sample, bins=bins, density=True, alpha=0.5, color=colors[i], edgecolor='white')

    kde = stats.gaussian_kde(sample)
    x_range = np.linspace(bins[0], bins[-1], 500)
    ax_top.plot(x_range, kde(x_range), color='darkblue', linewidth=2)

    # Истинное среднее
    ax_top.axvline(TRUE_MEAN, color='purple', linestyle='-', linewidth=2.5,
                   label=f'Истинное μ = {TRUE_MEAN}')

    # Выборочное среднее
    ax_top.axvline(mean_val, color='red', linestyle='--', linewidth=2,
                   label=f'x̄ = {mean_val:.2f}')

    # 95% ДИ
    ax_top.axvspan(ci_lower, ci_upper, alpha=0.25, color='red',
                   label=f'95% ДИ: [{ci_lower:.1f}, {ci_upper:.1f}]')

    # Подпись ширины ДИ
    ax_top.annotate(f'Ширина ДИ = {ci_upper - ci_lower:.1f}',
                    xy=((ci_lower+ci_upper)/2, ax_top.get_ylim()[1]*0.85),
                    ha='center', fontsize=10, color='darkred', fontweight='bold')

    ax_top.set_title(f'n = {n}', fontsize=13, fontweight='bold')
    ax_top.set_xlabel('Значение')
    if i == 0:
        ax_top.set_ylabel('Плотность')
    ax_top.legend(loc='upper left', fontsize=8.5)
    ax_top.grid(True, alpha=0.3)

    # ===== НИЗ: Box Plot =====
    ax_bot = axes[1, i]

    bp = ax_bot.boxplot(sample, vert=False, patch_artist=True, widths=0.4,
                        boxprops=dict(facecolor=colors[i], alpha=0.6, color='black', linewidth=1.5),
                        medianprops=dict(color='green', linewidth=2.5),
                        whiskerprops=dict(color='black', linewidth=1.5),
                        capprops=dict(color='black', linewidth=1.5),
                        flierprops=dict(marker='o', markerfacecolor='red', markersize=4, alpha=0.5))

    # Среднее
    ax_bot.plot(mean_val, 1, 'D', color='red', markersize=11, zorder=5)
    ax_bot.axvline(mean_val, color='red', linestyle='--', alpha=0.4)

    # Истинное среднее
    ax_bot.axvline(TRUE_MEAN, color='purple', linestyle='-', linewidth=2, alpha=0.7)
    ax_bot.plot(TRUE_MEAN, 1, '*', color='purple', markersize=13, zorder=5)

    # ДИ
    ax_bot.axvspan(ci_lower, ci_upper, alpha=0.2, color='red')
    ax_bot.annotate('95% ДИ\nсреднего', xy=((ci_lower+ci_upper)/2, 1.45),
                    ha='center', fontsize=10, color='darkred', fontweight='bold')

    # Подписи
    q1, q2, q3 = np.percentile(sample, [25, 50, 75])
    ax_bot.annotate(f'Q1={q1:.1f}', xy=(q1, 0.62), ha='center', fontsize=9, color='darkgreen')
    ax_bot.annotate(f'Медиана={q2:.1f}', xy=(q2, 0.52), ha='center', fontsize=9,
                    color='green', fontweight='bold')
    ax_bot.annotate(f'Q3={q3:.1f}', xy=(q3, 0.62), ha='center', fontsize=9, color='darkgreen')

    ax_bot.set_xlabel('Значение')
    ax_bot.set_yticks([1])
    ax_bot.set_yticklabels([''])
    ax_bot.grid(True, alpha=0.3, axis='x')

    if i == 0:
        ax_bot.legend(['Среднее', 'Истинное μ'], loc='lower right', fontsize=9)

# Пояснение
explanation = (
    "КЛЮЧЕВЫЕ НАБЛЮДЕНИЯ:\n\n"
    "• При n=30 ДИ широкий: мы мало знаем о генеральной совокупности, оценка среднего неточная.\n"
    "• При n=100 ДИ уже существенно сузился: больше данных = точнее оценка.\n"
    "• При n=500 ДИ очень узкий: выборочное среднее почти совпадает с истинным.\n\n"
    "Формула 95% ДИ:  x̄ ± t₀.₉₇₅ × SE,  где SE = s/√n  (стандартная ошибка среднего).\n"
    "При увеличении n в 4 раза ширина ДИ уменьшается примерно в 2 раза (обратно пропорционально √n).\n\n"
    "Box Plot показывает РАСПРЕДЕЛЕНИЕ данных (медиану, квартили, выбросы),\n"
    "а красная полоса (ДИ) — ТОЧНОСТЬ оценки СРЕДНЕГО. Это разные вещи!"
)

fig.text(0.5, -0.16, explanation, ha='center', fontsize=11,
         bbox=dict(boxstyle='round,pad=0.8', facecolor='lightyellow', alpha=0.95))

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('distribution_ci_boxplot_comparison.png', dpi=150, bbox_inches='tight', facecolor='white')
plt.show()
print("График сохранен!")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)

# --- Генерация данных ---
n = 150

# Контрольная группа
control = np.random.normal(100, 12, n)

# Экспериментальная группа (улучшение на ~8 единиц)
experimental = np.random.normal(108, 14, n)

# Статистики
m1, s1 = np.mean(control), np.std(control, ddof=1)
m2, s2 = np.mean(experimental), np.std(experimental, ddof=1)
se_diff = np.sqrt(s1**2/n + s2**2/n)
t_obs = (m2 - m1) / se_diff
df = (s1**2/n + s2**2/n)**2 / ((s1**2/n)**2/(n-1) + (s2**2/n)**2/(n-1))
p_value = 2 * (1 - stats.t.cdf(abs(t_obs), df))

# 95% ДИ для разности
ci_diff_lower = (m2 - m1) - stats.t.ppf(0.975, df) * se_diff
ci_diff_upper = (m2 - m1) + stats.t.ppf(0.975, df) * se_diff

print("=" * 55)
print("СРАВНЕНИЕ КОНТРОЛЬНОЙ И ЭКСПЕРИМЕНТАЛЬНОЙ ГРУПП")
print("=" * 55)
print(f"Контрольная:       n={n}, mean={m1:.2f}, std={s1:.2f}")
print(f"Экспериментальная: n={n}, mean={m2:.2f}, std={s2:.2f}")
print(f"Разность (Exp - Ctrl): {m2 - m1:.2f}")
print(f"95% ДИ разности:     [{ci_diff_lower:.2f}, {ci_diff_upper:.2f}]")
print(f"t = {t_obs:.3f}, df = {df:.1f}, p-value = {p_value:.4f}")
print("=" * 55)

# --- ВИЗУАЛИЗАЦИЯ ---
fig = plt.figure(figsize=(16, 13))
gs = fig.add_gridspec(3, 3, height_ratios=[1.3, 1, 1.1], hspace=0.38, wspace=0.3)

# ===== ПАНЕЛЬ 1: Гистограммы с наложением =====
ax1 = fig.add_subplot(gs[0, :])

bins = np.linspace(min(control.min(), experimental.min()) - 5,
                   max(control.max(), experimental.max()) + 5, 40)

ax1.hist(control, bins=bins, alpha=0.5, color='steelblue', edgecolor='white',
         label=f'Контрольная (mean={m1:.1f})', density=True)
ax1.hist(experimental, bins=bins, alpha=0.5, color='coral', edgecolor='white',
         label=f'Экспериментальная (mean={m2:.1f})', density=True)

x_range = np.linspace(bins[0], bins[-1], 500)
kde1 = stats.gaussian_kde(control)
kde2 = stats.gaussian_kde(experimental)
ax1.plot(x_range, kde1(x_range), color='darkblue', linewidth=2.5)
ax1.plot(x_range, kde2(x_range), color='darkred', linewidth=2.5)

ax1.axvline(m1, color='steelblue', linestyle='--', linewidth=2.5)
ax1.axvline(m2, color='coral', linestyle='--', linewidth=2.5)

ax1.annotate(f'Контрольная\nmean = {m1:.1f}', xy=(m1, ax1.get_ylim()[1]*0.65),
             ha='center', fontsize=11, color='steelblue', fontweight='bold',
             bbox=dict(boxstyle='round,pad=0.4', facecolor='lightblue', alpha=0.7))
ax1.annotate(f'Экспериментальная\nmean = {m2:.1f}', xy=(m2, ax1.get_ylim()[1]*0.65),
             ha='center', fontsize=11, color='coral', fontweight='bold',
             bbox=dict(boxstyle='round,pad=0.4', facecolor='lightsalmon', alpha=0.7))

ax1.annotate('', xy=(m2, ax1.get_ylim()[1]*0.3), xytext=(m1, ax1.get_ylim()[1]*0.3),
             arrowprops=dict(arrowstyle='<->', color='green', lw=2.5))
ax1.text((m1+m2)/2, ax1.get_ylim()[1]*0.38, f'Δ = {m2-m1:.1f}\np = {p_value:.4f}',
         ha='center', fontsize=12, color='darkgreen', fontweight='bold',
         bbox=dict(boxstyle='round,pad=0.4', facecolor='lightgreen', alpha=0.85))

ax1.set_title('Распределения данных в группах', fontsize=14, fontweight='bold', pad=10)
ax1.set_xlabel('начение', fontsize=11)
ax1.set_ylabel('Плотность', fontsize=11)
ax1.legend(loc='upper left', fontsize=10)
ax1.grid(True, alpha=0.3)

# ===== ПАНЕЛЬ 2: Box Plot =====
ax2 = fig.add_subplot(gs[1, 0])

bp = ax2.boxplot([control, experimental], vert=False, patch_artist=True, widths=0.5,
                 labels=['Контрольная', 'Экспериментальная'],
                 boxprops=dict(linewidth=1.5),
                 medianprops=dict(color='green', linewidth=2.5),
                 whiskerprops=dict(linewidth=1.5),
                 capprops=dict(linewidth=1.5),
                 flierprops=dict(marker='o', markersize=5, alpha=0.5))

for patch, color in zip(bp['boxes'], ['steelblue', 'coral']):
    patch.set_facecolor(color)
    patch.set_alpha(0.6)

ax2.plot([m1, m2], [1, 2], 'D', color='red', markersize=11, label='Среднее')

q1_c, med_c, q3_c = np.percentile(control, [25, 50, 75])
q1_e, med_e, q3_e = np.percentile(experimental, [25, 50, 75])
ax2.annotate(f'Q1={q1_c:.1f}', xy=(q1_c, 0.65), ha='center', fontsize=9, color='darkgreen')
ax2.annotate(f'Мед={med_c:.1f}', xy=(med_c, 0.55), ha='center', fontsize=9, color='green', fontweight='bold')
ax2.annotate(f'Q3={q3_c:.1f}', xy=(q3_c, 0.65), ha='center', fontsize=9, color='darkgreen')
ax2.annotate(f'Q1={q1_e:.1f}', xy=(q1_e, 1.65), ha='center', fontsize=9, color='darkgreen')
ax2.annotate(f'Мед={med_e:.1f}', xy=(med_e, 1.55), ha='center', fontsize=9, color='green', fontweight='bold')
ax2.annotate(f'Q3={q3_e:.1f}', xy=(q3_e, 1.65), ha='center', fontsize=9, color='darkgreen')

ax2.set_title('Box Plot', fontsize=13, fontweight='bold')
ax2.set_xlabel('Значение', fontsize=11)
ax2.legend(loc='lower right', fontsize=9)
ax2.grid(True, alpha=0.3, axis='x')

# ===== ПАНЕЛЬ 3: Violin Plot =====
ax3 = fig.add_subplot(gs[1, 1])

parts = ax3.violinplot([control, experimental], positions=[1, 2],
                       showmeans=False, showmedians=False, showextrema=False, widths=0.7)

for pc, color in zip(parts['bodies'], ['steelblue', 'coral']):
    pc.set_facecolor(color)
    pc.set_alpha(0.6)
    pc.set_edgecolor('black')
    pc.set_linewidth(1.5)

# Box plot внутри violin
ax3.boxplot([control, experimental], positions=[1, 2], widths=0.12,
            showfliers=False, patch_artist=True,
            boxprops=dict(facecolor='white', alpha=0.8, linewidth=1),
            medianprops=dict(color='green', linewidth=2),
            whiskerprops=dict(linewidth=1),
            capprops=dict(linewidth=1))

ax3.plot([1, 2], [m1, m2], 'D', color='red', markersize=10, label='Среднее')
ax3.plot([1, 2], [med_c, med_e], 's', color='green', markersize=8, label='Медиана')

ax3.set_yticks([1, 2])
ax3.set_yticklabels(['Контрольная', 'Экспериментальная'])
ax3.set_title('Violin Plot (+ Box)', fontsize=13, fontweight='bold')
ax3.set_xlabel('Значение', fontsize=11)
ax3.legend(loc='lower right', fontsize=9)
ax3.grid(True, alpha=0.3, axis='x')

# ===== ПАНЕЛЬ 4: Разделенный Violin Plot (вручную через KDE) =====
ax4 = fig.add_subplot(gs[1, 2])

# Вычисляем KDE для обеих групп
kde_c = stats.gaussian_kde(control)
kde_e = stats.gaussian_kde(experimental)

y_range = np.linspace(min(control.min(), experimental.min()) - 5,
                      max(control.max(), experimental.max()) + 5, 500)

density_c = kde_c(y_range)
density_e = kde_e(y_range)

# Нормализуем для красоты
max_d = max(density_c.max(), density_e.max())
density_c = density_c / max_d * 0.4
density_e = density_e / max_d * 0.4

# Рисуем половинки
ax4.fill_betweenx(y_range, -density_c, 0, color='steelblue', alpha=0.6, label='Контрольная')
ax4.fill_betweenx(y_range, 0, density_e, color='coral', alpha=0.6, label='Экспериментальная')

# Средние и медианы
ax4.plot(-0.05, m1, 'D', color='steelblue', markersize=11)
ax4.plot(0.05, m2, 'D', color='coral', markersize=11)
ax4.plot(-0.05, med_c, 's', color='green', markersize=8)
ax4.plot(0.05, med_e, 's', color='green', markersize=8)

# Линии средних
ax4.axhline(m1, color='steelblue', linestyle='--', alpha=0.4, xmax=0.5)
ax4.axhline(m2, color='coral', linestyle='--', alpha=0.4, xmin=0.5)

# Линия раздела
ax4.axvline(0, color='black', linewidth=1)

# Легенда
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], marker='D', color='w', markerfacecolor='steelblue', markersize=10, label='Среднее (Ctrl)'),
    Line2D([0], [0], marker='D', color='w', markerfacecolor='coral', markersize=10, label='Среднее (Exp)'),
    Line2D([0], [0], marker='s', color='w', markerfacecolor='green', markersize=8, label='Медиана')
]
ax4.legend(handles=legend_elements, loc='upper right', fontsize=9)

ax4.set_xlim(-0.5, 0.5)
ax4.set_xticks([0])
ax4.set_xticklabels(['Сранение\nгрупп'])
ax4.set_title('Разделенный Violin Plot', fontsize=13, fontweight='bold')
ax4.set_ylabel('Значение', fontsize=11)
ax4.grid(True, alpha=0.3, axis='y')

# ===== ПАНЕЛЬ 5: Средние с доверительными интервалами =====
ax5 = fig.add_subplot(gs[2, :])

se1 = s1 / np.sqrt(n)
se2 = s2 / np.sqrt(n)
ci1 = stats.t.interval(0.95, n-1, loc=m1, scale=se1)
ci2 = stats.t.interval(0.95, n-1, loc=m2, scale=se2)

y_pos = [1, 2]
means = [m1, m2]
errors_lower = [m1 - ci1[0], m2 - ci2[0]]
errors_upper = [ci1[1] - m1, ci2[1] - m2]
bar_colors = ['steelblue', 'coral']

ax5.barh(y_pos, means, xerr=[errors_lower, errors_upper],
         color=bar_colors, alpha=0.7, capsize=10, edgecolor='black', linewidth=1.5, height=0.5)

ax5.set_yticks(y_pos)
ax5.set_yticklabels(['Контрольная', 'Экспериментальная'])
ax5.set_xlabel('Значение', fontsize=11)
ax5.set_title('Средние значения с 95% доверительными интервалами', fontsize=13, fontweight='bold')

for i, (m, ci_l, ci_u) in enumerate(zip(means, [ci1[0], ci2[0]], [ci1[1], ci2[1]])):
    ax5.text(m + 2, i + 1.18, f'{m:.2f}\n95% ДИ: [{ci_l:.1f}, {ci_u:.1f}]',
             ha='center', fontsize=10, color=bar_colors[i], fontweight='bold')

ax5.annotate('', xy=(m2, 1.3), xytext=(m1, 1.3),
             arrowprops=dict(arrowstyle='<->', color='green', lw=2.5))
ax5.text((m1+m2)/2, 1.48,
         f'Разность = {m2-m1:.2f}\n95% ДИ разности: [{ci_diff_lower:.1f}, {ci_diff_upper:.1f}]\np-value = {p_value:.4f}',
         ha='center', fontsize=11, color='darkgreen', fontweight='bold',
         bbox=dict(boxstyle='round,pad=0.5', facecolor='lightgreen', alpha=0.85))

ax5.grid(True, alpha=0.3, axis='x')
ax5.set_xlim(min(ci1[0], ci2[0]) - 5, max(ci1[1], ci2[1]) + 15)

plt.suptitle('Наглядное сравнение контрольной и экспериментальной групп',
             fontsize=16, fontweight='bold', y=1.01)

plt.savefig('control_vs_experiment.png', dpi=150, bbox_inches='tight', facecolor='white')
plt.show()
print("График сохранен!")
```

# В дополнение к вероятностным методам
для анализа тренировок лучше всего подходят следующие разделы: 1. Теория нечетких множеств и нечеткая логика (Fuzzy Logic)Идеально подходит для оцифровки субъективных ощущений спортсмена (RPE — субъективный уровень нагрузки) и качественных показателей. Как это работает: Вместо жестких границ (например, пульс >160 — это «высокий») вводятся функции принадлежности. Пульс 158 может быть на 80% «высоким» и на 20% «умеренным». Применение:Автоматический расчет готовности к тренировке на основе анкет («как спалось», «какой уровень стресса»).Определение зон интенсивности, которые плавно переходят одна в другую из-за ежедневных колебаний формы. 2. Интервальный анализПрименяется для работы с погрешностями спортивных гаджетов (GPS-часов, пульсометров, датчиков мощности). Как это работает: Данные с трекеров воспринимаются не как точные точки, а как коридоры значений. Например, текущий темп бега не ровно 4:30 мин/км, а интервал \([4:25, 4:35]\) из-за погрешности GPS. Применение:Фильтрация аномалий и «скачков» пульса или мощности (когда датчик на секунду потерял контакт с кожей).Корректный расчет зон интенсивности с учетом аппаратных допусков устройств. 3.
```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt

# =====================================================================
# 1. ОПРЕДЕЛЕНИЕ ПЕРЕМЕННЫХ (Входы и Выход)
# =====================================================================

# Входы:
# hrv (ВСР) — показатель восстановления (от 0 до 100). Выше — лучше.
# fatigue (Усталость) — субъективная оценка от 0 (бодр) до 10 (истощен).
hrv = ctrl.Antecedent(np.arange(0, 101, 1), 'hrv')
fatigue = ctrl.Antecedent(np.arange(0, 11, 1), 'fatigue')

# Выход:
# load_mod (Коррекция нагрузки) — от -50% (разгрузка) до +20% (интенсивный тренинг)
load_mod = ctrl.Consequent(np.arange(-50, 21, 1), 'load_mod')

# =====================================================================
# 2. ФУНКЦИИ ПРИНАДЛЕЖНОСТИ (Оцифровка понятий)
# =====================================================================

# Для ВСР (HRV): низкий, нормальный, отличный
hrv['low'] = fuzz.trapmf(hrv.universe, [0, 0, 35, 50])
hrv['normal'] = fuzz.trimf(hrv.universe, [40, 60, 80])
hrv['excellent'] = fuzz.trapmf(hrv.universe, [70, 85, 100, 100])

# Для усталости (Fatigue): низкая, средняя, высокая
fatigue['low'] = fuzz.trimf(fatigue.universe, [0, 0, 4])
fatigue['medium'] = fuzz.trimf(fatigue.universe, [3, 5, 7])
fatigue['high'] = fuzz.trapmf(fatigue.universe, [6, 8, 10, 10])

# Для корректировки нагрузки: Сбросить (Drop), Сохранить (Keep), Добавить (Boost)
load_mod['drop'] = fuzz.trapmf(load_mod.universe, [-50, -50, -25, 0])
load_mod['keep'] = fuzz.trimf(load_mod.universe, [-15, 0, 10])
load_mod['boost'] = fuzz.trimf(load_mod.universe, [0, 20, 20])


hrv.view()

# =====================================================================
# 3. НЕЧЁТКИЕ ПРАВИЛА (База знаний тренера)
# =====================================================================
rule1 = ctrl.Rule(hrv['low'] | fatigue['high'], load_mod['drop'])
rule2 = ctrl.Rule(hrv['normal'] & fatigue['medium'], load_mod['keep'])
rule3 = ctrl.Rule(hrv['excellent'] & fatigue['low'], load_mod['boost'])
rule4 = ctrl.Rule(hrv['normal'] & fatigue['low'], load_mod['keep'])
rule5 = ctrl.Rule(hrv['excellent'] & fatigue['medium'], load_mod['keep'])

# Создаем управляющую систему
training_control = ctrl.ControlSystem([rule1, rule2, rule3, rule4, rule5])
training_simulator = ctrl.ControlSystemSimulation(training_control)

# =====================================================================
# 4. РАСЧЕТ ДЛЯ КОНКРЕТНОГО СПОРТСМЕНА
# =====================================================================

# Пример: Спортсмен невыспался, усталость 7 из 10, но ВСР еще в норме (55).
current_hrv = 55
current_fatigue = 7

training_simulator.input['hrv'] = current_hrv
training_simulator.input['fatigue'] = current_fatigue

# Запуск вычислений
training_simulator.compute()

# Получение четкого (дефаззифицированного) результата
recommended_change = training_simulator.output['load_mod']

print(f"--- РЕЗУЛЬТАТ АНАЛИЗА ---")
print(f"Входные данные -> ВСР: {current_hrv}, Усталость: {current_fatigue}/10")
print(f"Рекомендация по изменению нагрузки: {recommended_change:.1f}%")

# Визуализация результата дефаззификации прямо в Colab
load_mod.view(sim=training_simulator)
plt.show()

```

1. Заполнение данных: Код принимает на вход жесткие числа (ВСР = 55, Усталость = 7).
2. Фаззификация: Алгоритм переводит их в степени истинности. Например, усталость 7 — это одновременно «немного средняя» и «уже высокая».
3. Логический вывод: Система активирует встроенные правила тренера (например, если усталость высокая, надо сбросить нагрузку).
4. Дефаззификация: Магия библиотеки skfuzzy собирает пересекающиеся графики правил и методом центроида вычисляет одну точную цифру. В данном примере программа выдаст рекомендацию снизить нагрузку примерно на -20.8% и нарисует график, показывающий, какая часть зоны drop (сброс) была задействована.

Весь процесс нечеткой логики состоит из 3 шагов:
1. Фаззификация («Размытие» чисел):
Робот не понимает фразу «усталость 7». Код берет это число и смотрит на графики. Он видит, что 7 попадает одновременно в два графика: это на 40% «Средняя усталость» и на 60% «Высокая усталость». Жесткое число превратилось в степени уверенности.
2. Логический вывод (Применение правил):
Код берет правила, которые мы прописали (базу знаний тренера). Например: «Если усталость ВЫСОКАЯ, то нагрузку надо СБРОСИТЬ». Раз мы уверены в высокой усталости на 60%, то и правило «СБРОСИТЬ нагрузку» активируется на 60%. График срезается по этой высоте.
3. Дефаззификация («Сборка» обратно в число):
Код накладывает все активированные графики решений друг на друга. Получается сложная геометрическая фигура. Алгоритм находит центр тяжести (центроид) этой фигуры. Координата этого центра на оси процентов и есть итоговый ответ (например, -20.8%).



с помощью библиотеки scikit-fuzzy решается классическая задача тренера: расчет рекомендаций по нагрузке на сегодня на основе двух факторов — уровня автономного стресса (вариабельности сердечного ритма, ВСР) и субъективного ощущения усталости спортсмена.


```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# =====================================================================
# 1. НАШИ РЕАЛЬНЫЕ ДАННЫЕ (Исторический опыт спортсмена)
# Строка: [Значение_ВСР, Значение_Усталости, Коррекция_Нагрузки_%]
# =====================================================================
historical_data = np.array([
    [80, 2, 15],   # Высокий ВСР, усталость 2 из 10 -> добавил +15%
    [30, 9, -40],  # Низкий ВСР, усталость 9 из 10 -> снизил на -40%
    [55, 5, 0],    # Средние показатели -> нагрузка без изменений (0%)
    [70, 3, 10],   # Хорошее состояние -> чуть добавил +10%
    [40, 8, -30]   # Сильная усталость -> снизил на -30%
])

# =====================================================================
# 2. ФУНКЦИЯ СОЗДАНИЯ СИСТЕМЫ И РАСЧЕТА ОШИБКИ
# =====================================================================
def evaluate_fuzzy_system(params):
    # Распаковываем 4 точки (центры графиков)
    hrv_norm_center, fatigue_med_center, load_keep_center, load_boost_center = params

    # Сортируем или жестко страхуем геометрию, чтобы a <= b <= c всегда выполнялось
    # Для load_mod: у нас есть точки -20, load_keep_center, load_boost_center, 20
    # Нам нужно убедиться, что -20 <= load_keep_center <= load_boost_center <= 20
    if not (-20 <= load_keep_center <= load_boost_center <= 20):
        return 1e7  # Огромный штраф, если порядок нарушен

    if not (30 <= hrv_norm_center <= 80) or not (2 <= fatigue_med_center <= 8):
        return 1e7

    # Создаем шкалы
    hrv = ctrl.Antecedent(np.arange(0, 101, 1), 'hrv')
    fatigue = ctrl.Antecedent(np.arange(0, 11, 1), 'fatigue')
    load_mod = ctrl.Consequent(np.arange(-50, 21, 1), 'load_mod')

    # Настраиваем графики с защитой геометрии
    hrv['low'] = fuzz.trapmf(hrv.universe, [0, 0, 30, hrv_norm_center])
    hrv['normal'] = fuzz.trimf(hrv.universe, [30, hrv_norm_center, 80])
    hrv['excellent'] = fuzz.trapmf(hrv.universe, [hrv_norm_center, 80, 100, 100])

    fatigue['low'] = fuzz.trimf(fatigue.universe, [0, 0, fatigue_med_center])
    fatigue['medium'] = fuzz.trimf(fatigue.universe, [2, fatigue_med_center, 8])
    fatigue['high'] = fuzz.trapmf(fatigue.universe, [fatigue_med_center, 8, 10, 10])

    load_mod['drop'] = fuzz.trapmf(load_mod.universe, [-50, -50, -25, load_keep_center])
    load_mod['keep'] = fuzz.trimf(load_mod.universe, [-20, load_keep_center, load_boost_center])
    load_mod['boost'] = fuzz.trimf(load_mod.universe, [load_keep_center, load_boost_center, 20])

    # Базовые правила логики
    rule1 = ctrl.Rule(hrv['low'] | fatigue['high'], load_mod['drop'])
    rule2 = ctrl.Rule(hrv['normal'] & fatigue['medium'], load_mod['keep'])
    rule3 = ctrl.Rule(hrv['excellent'] & fatigue['low'], load_mod['boost'])

    training_control = ctrl.ControlSystem([rule1, rule2, rule3])
    sim = ctrl.ControlSystemSimulation(training_control)

    # Считаем сумму квадратов ошибок
    total_error = 0
    for row in historical_data:
        sim.input['hrv'] = row[0]
        sim.input['fatigue'] = row[1]
        try:
            sim.compute()
            model_output = sim.output['load_mod']
            total_error += (model_output - row[2]) ** 2
        except:
            total_error += 10000

    return total_error

# =====================================================================
# 3. ЗАПУСК ОПТИМИЗАЦИИ С ОГРАНИЧЕНИЯМИ (BOUNDS)
# =====================================================================
print("Запуск автоподбора параметров по историческим данным...")

# Начальные догадки [hrv_norm, fatigue_med, load_keep, load_boost]
initial_guess = [50.0, 5.0, 0.0, 10.0]

# Задаем жесткие рамки для каждого параметра (min, max), чтобы они не путались местами
bounds = [
    (35.0, 75.0),   # hrv_norm_center
    (3.0, 7.0),     # fatigue_med_center
    (-15.0, 5.0),   # load_keep_center (должен быть меньше, чем load_boost)
    (6.0, 18.0)     # load_boost_center
]

# Используем метод L-BFGS-B, который строго уважает границы bounds
result = minimize(evaluate_fuzzy_system, initial_guess, method='L-BFGS-B', bounds=bounds)

best_params = result.x
print("\n--- ОПТИМИЗАЦИЯ СИСТЕМЫ ЗАВЕРШЕНА ---")
print(f"Идеальный центр для 'Нормального ВСР': {best_params[0]:.1f}")
print(f"Идеальный центр для 'Средней усталости': {best_params[1]:.1f}")
print(f"Идеальная точка сохранения нагрузки (0%): {best_params[2]:.1f}%")
print(f"Идеальный пик для зоны добавления нагрузки (+10%): {best_params[3]:.1f}%")
```

```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# =====================================================================
# 1. ИСТОРИЧЕСКИЕ ДАННЫЕ
# =====================================================================
historical_data = np.array([
    [80, 2, 15],   # Высокий ВСР, усталость 2 из 10 -> добавил +15%
    [30, 9, -40],  # Низкий ВСР, усталость 9 из 10 -> снизил на -40%
    [55, 5, 0],    # Средние показатели -> нагрузка без изменений (0%)
    [70, 3, 10],   # Хорошее состояние -> чуть добавил +10%
    [40, 8, -30]   # Сильная усталость -> снизил на -30%
])

# =====================================================================
# 2. ФУНКЦИЯ ОПТИМИЗАЦИИ (БЕЗ ИЗМЕНЕНИЙ, ДЛЯ РАСЧЕТА)
# =====================================================================
def evaluate_fuzzy_system(params):
    hrv_norm_center, fatigue_med_center, load_keep_center, load_boost_center = params
    if not (-20 <= load_keep_center <= load_boost_center <= 20): return 1e7
    if not (30 <= hrv_norm_center <= 80) or not (2 <= fatigue_med_center <= 8): return 1e7

    hrv = ctrl.Antecedent(np.arange(0, 101, 1), 'hrv')
    fatigue = ctrl.Antecedent(np.arange(0, 11, 1), 'fatigue')
    load_mod = ctrl.Consequent(np.arange(-50, 21, 1), 'load_mod')

    hrv['low'] = fuzz.trapmf(hrv.universe, [0, 0, 30, hrv_norm_center])
    hrv['normal'] = fuzz.trimf(hrv.universe, [30, hrv_norm_center, 80])
    hrv['excellent'] = fuzz.trapmf(hrv.universe, [hrv_norm_center, 80, 100, 100])

    fatigue['low'] = fuzz.trimf(fatigue.universe, [0, 0, fatigue_med_center])
    fatigue['medium'] = fuzz.trimf(fatigue.universe, [2, fatigue_med_center, 8])
    fatigue['high'] = fuzz.trapmf(fatigue.universe, [fatigue_med_center, 8, 10, 10])

    load_mod['drop'] = fuzz.trapmf(load_mod.universe, [-50, -50, -25, load_keep_center])
    load_mod['keep'] = fuzz.trimf(load_mod.universe, [-20, load_keep_center, load_boost_center])
    load_mod['boost'] = fuzz.trimf(load_mod.universe, [load_keep_center, load_boost_center, 20])

    rule1 = ctrl.Rule(hrv['low'] | fatigue['high'], load_mod['drop'])
    rule2 = ctrl.Rule(hrv['normal'] & fatigue['medium'], load_mod['keep'])
    rule3 = ctrl.Rule(hrv['excellent'] & fatigue['low'], load_mod['boost'])

    training_control = ctrl.ControlSystem([rule1, rule2, rule3])
    sim = ctrl.ControlSystemSimulation(training_control)

    total_error = 0
    for row in historical_data:
        sim.input['hrv'] = row[0]
        sim.input['fatigue'] = row[1]
        try:
            sim.compute()
            total_error += (sim.output['load_mod'] - row[2]) ** 2
        except:
            total_error += 10000
    return total_error

# Запуск оптимизации
# Начальные догадки [hrv_norm, fatigue_med, load_keep, load_boost]
initial_guess = [50.0, 5.0, 0.0, 10.0]

# Задаем жесткие рамки для каждого параметра (min, max), чтобы они не путались местами
bounds = [
    (35.0, 75.0),   # hrv_norm_center
    (3.0, 7.0),     # fatigue_med_center
    (-15.0, 5.0),   # load_keep_center (должен быть меньше, чем load_boost)
    (6.0, 18.0)     # load_boost_center
]

# Используем метод L-BFGS-B, который строго уважает границы bounds
result = minimize(evaluate_fuzzy_system, initial_guess, method='L-BFGS-B', bounds=bounds)

best_params = result.x
print("\n--- ОПТИМИЗАЦИЯ СИСТЕМЫ ЗАВЕРШЕНА ---")
print(f"Идеальный центр для 'Нормального ВСР': {best_params[0]:.1f}")
print(f"Идеальный центр для 'Средней усталости': {best_params[1]:.1f}")
print(f"Идеальная точка сохранения нагрузки (0%): {best_params[2]:.1f}%")
print(f"Идеальный пик для зоны добавления нагрузки (+10%): {best_params[3]:.1f}%")

# =====================================================================
# 3. ПОСТРОЕНИЕ ПОНЯТНЫХ ГРАФИКОВ С НОВЫМИ ПАРАМЕТРАМИ
# =====================================================================
h_cent, f_cent, l_keep, l_boost = best_params

# Создаем финальные массивы для красивой отрисовки
x_hrv = np.arange(0, 101, 1)
x_fatigue = np.arange(0, 11, 1)
x_load = np.arange(-50, 21, 1)

plt.figure(figsize=(15, 4))

# График 1: ВСР
plt.subplot(1, 3, 1)
plt.plot(x_hrv, fuzz.trapmf(x_hrv, [0, 0, 30, h_cent]), label='Низкий', color='red')
plt.plot(x_hrv, fuzz.trimf(x_hrv, [30, h_cent, 80]), label='Норма', color='orange')
plt.plot(x_hrv, fuzz.trapmf(x_hrv, [h_cent, 80, 100, 100]), label='Отличный', color='green')
plt.axvline(h_cent, color='blue', linestyle='--', alpha=0.7)
plt.title(f'ВСР (Идеальная норма: {h_cent:.1f})')
plt.xlabel('Показатель ВСР')
plt.ylabel('Степень уверенности (0-1)')
plt.legend()

# График 2: Усталость
plt.subplot(1, 3, 2)
plt.plot(x_fatigue, fuzz.trimf(x_fatigue, [0, 0, f_cent]), label='Низкая', color='green')
plt.plot(x_fatigue, fuzz.trimf(x_fatigue, [2, f_cent, 8]), label='Средняя', color='orange')
plt.plot(x_fatigue, fuzz.trapmf(x_fatigue, [f_cent, 8, 10, 10]), label='Высокая', color='red')
plt.axvline(f_cent, color='blue', linestyle='--', alpha=0.7)
plt.title(f'Усталость (Пик средней: {f_cent:.1f})')
plt.xlabel('Субъективная оценка (0-10)')
plt.legend()

# График 3: Нагрузка
plt.subplot(1, 3, 3)
plt.plot(x_load, fuzz.trapmf(x_load, [-50, -50, -25, l_keep]), label='Сбросить', color='red')
plt.plot(x_load, fuzz.trimf(x_load, [-20, l_keep, l_boost]), label='Сохранить', color='orange')
plt.plot(x_load, fuzz.trimf(x_load, [l_keep, l_boost, 20]), label='Добавить', color='green')
plt.axvline(l_keep, color='blue', linestyle='--', alpha=0.5)
plt.axvline(l_boost, color='blue', linestyle='--', alpha=0.5)
plt.title(f'Коррекция (Плата: {l_keep:.1f}%, Пик: +{l_boost:.1f}%)')
plt.xlabel('Изменение нагрузки в %')
plt.legend()

plt.tight_layout()
plt.show()

# =====================================================================
# 4. ТЕКСТОВОЕ ОБЪЯСНЕНИЕ РЕЗУЛЬТАТОВ
# =====================================================================
print("\n" + "="*70)
print("РАЗБОР НАСТРОЕК НЕЧЕТКОЙ СИСТЕМЫ")
print("="*70)
print(f"1. ГРАФИК ВСР: Вершина 'Нормального ВСР' сдвинулась в точку {h_cent:.1f}.")
print(f"   Интерпретация: Индивидуальный организм этого атлета считает балансом именно этот пульс.")
print(f"   Всё, что ниже 30 — это гарантированный недовосстановленный спад ('Низкий ВСР').")
print(f"\n2. ГРАФИК УСТАЛОСТИ: Центр 'Средней усталости' зафиксирован на отметке {f_cent:.1f} из 10.")
print(f"   Интерпретация: Когда спортсмен ставит себе {f_cent:.1f}, система на 100% уверена, что")
print(f"   нагрузка должна оставаться базовой. Выше этой точки быстро растет зона риска ('Высокая').")
print(f"\n3. РЕКОМЕНДАЦИИ ПО НАГРУЗКЕ:")
print(f"   - Точка стабильности ('Сохранить'): {l_keep:.1f}%.")
print(f"     При нормальном самочувствии система рекомендует держать нагрузку около этой цифры.")
print(f"   - Пик прогресса ('Добавить'): +{l_boost:.1f}%.")
print(f"     Когда показатели идеальны, оптимизатор вычислил, что безопасный и эффективный")
print(f"     максимум для добавления рабочего веса или объема составляет именно {l_boost:.1f}%, а не +10%.")
print("="*70)
```

1. Первый график (ВСР): Вы увидите, где пересекаются цветные линии. Точка, где оранжевый треугольник («Норма») достигает самой верхней отметки (единицы) — это и есть вычисленный идеальный баланс ВСР для конкретного человека.
2. Второй график (Усталость): Вершина оранжевого треугольника показывает субъективный балл, который является чистым «экватором» состояния. Справа от него начинает резко расти красная линия — зона «Высокой усталости», где включается правило экстренного сброса нагрузок.
3. Третий график (Нагрузка): Показывает зоны решений. Вершины треугольников здесь определяют, к какому именно проценту (центру тяжести) будет стремиться итоговое решение алгоритма при хорошем или среднем состоянии.


Выбор начальных параметров (initial guess) и границ (bounds) — это важнейший этап, который превращает абстрактную математическую модель в работающий инструмент для конкретного вида спорта.
В реальной практике эти значения определяются на стыке двух подходов: спортивной науки (физиологии) и экспертного опыта тренера.
1. Как определяются начальные параметры (Initial Guess)
Начальные параметры — это ваша «стартовая гипотеза». Оптимизатору нужна точка отсчета, от которой он начнет двигаться. Если задать их слишком далеко от реальности, алгоритм может застрять в локальном минимуме или вообще не сойтись.
Они выставляются на основе усредненных медианных значений для выбранной метрики:
• Для субъективных шкал (Усталость, Стресс, Сон): Используется чистый экватор шкалы. Если шкала от 0 до 10, то начальной точкой для «Средней усталости» логично поставить 5.0.
• Для физиологических шкал (ВСР, ЧСС): Берутся средние нормы для популяции или конкретной возрастной группы. Для ВСР (в зависимости от метрики, например, rMSSD) норма здорового человека часто колеблется в районе 40–60 мс. Поэтому 50.0 — идеальный стартовый ориентир.
• Для шкал управления (Изменение нагрузки): Здесь закладывается классическая логика периодизации. По умолчанию при нормальном состоянии нагрузку менять не нужно (0.0%), а при отличном состоянии стандартный шаг микроцикла для линейного прогресса составляет около +10.0%.
2. Как определяются границы (Bounds)
Границы выполняют роль «предохранителя» для алгоритма. Они не позволяют математике принимать абсурдные или опасные для здоровья спортсмена решения (например, приказать добавить +150% к нагрузке или сдвинуть зону «нормального пульса» в диапазон, где у человека наступает тахикардия).
Границы рассчитываются по жестким правилам:
А. На основе физиологических лимитов человека
Вы не можете поставить нижнюю границу пульса в 0, а верхнюю в 300.
• Пример для ВСР: Мы знаем, что даже у профессиональных атлетов в состоянии глубокого переутомления ВСР редко падает ниже 20–25 мс, а в пике формы редко уходит за 100–120 мс. Значит, диапазон поиска «индивидуальной нормы» для оптимизатора имеет смысл сузить до рамок (35.0, 75.0).
Б. На основе теории и методики спортивной тренировки (Протоколы безопасности)
В спорте высших достижений есть жесткие лимиты на прирост объемов (например, правило 10% в беге, которое гласит, что нельзя увеличивать недельный метраж более чем на 10% без риска травмы).
• Пример для нагрузки: Как бы оптимизатор ни хотел подогнать модель под исторические данные, тренер обязан выставить жесткую верхнюю границу для зоны load_boost_center (например, не более 18.0% или 20.0%). Модель физически не сможет порекомендовать прыгнуть выше этого безопасного потолка.
В. На основе математической логики графиков
Границы не должны позволять треугольникам «наползать» друг на друга так, чтобы ломался их физический смысл.
• Пример: Пик «Сохранить нагрузку» (load_keep_center) всегда должен быть строго меньше, чем пик «Добавить нагрузку» (load_boost_center). Поэтому для первого мы задаем диапазон (-15.0, 5.0), а для второго — (6.0, 18.0). Их границы не пересекаются, и геометрия модели остается чистой.
Резюме: Алгоритм действий для новой метрики
Если вы захотите добавить в модель новый параметр (например, Продолжительность сна в часах), вы определите параметры так:
1. Метрика: Время сна (в часах).
2. Initial Guess: 7.5 часов (золотой стандарт здорового сна).
3. Bounds: (6.0, 9.0). Оптимизатор будет искать индивидуальный оптимум спортсмена внутри этого диапазона. Искать его на отметке 4 часа или 12 часов бессмысленно — это в любом случае аномалии, которые ведут к деградации формы.


1. Появился четвертый график (Сон): Синяя линия на первом графике показывает вычисленную точку s_cent (например, 7.2 часа). Это индивидуальный «рубикон».
2. Взаимосвязь параметров: Теперь правила работают по принципу логического «ИЛИ» на негативные факторы. Если спортсмен поставил хорошую оценку усталости (2.0), у него отличный ВСР (75), но он спал всего 4 часа — система проигнорирует хорошие показатели и выдаст жесткую рекомендацию drop (сбросить нагрузку), защищая сердце и связки.


```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# =====================================================================
# 1. ОБНОВЛЕННЫЕ ИСТОРИЧЕСКИЕ ДАННЫЕ (Добавлен 4-й столбец: Сон в часах)
# Строка: [ВСР, Усталость, Сон(часы), Коррекция_Нагрузки_%]
# =====================================================================
historical_data = np.array([
    [75, 2, 8.5,  15],  # Свеж, выспался (8.5ч) -> Нагрузка +15%
    [30, 9, 4.5, -40],  # Устал, не спал (4.5ч) -> Нагрузка -40%
    [50, 5, 7.0,   0],  # Средне, поспал нормально (7ч) -> 0%
    [65, 3, 8.0,  10],  # Хорошо, поспал 8ч -> Нагрузка +10%
    [40, 8, 5.5, -30]   # Плохо, недосып (5.5ч) -> Нагрузка -30%
])

# =====================================================================
# 2. ФУНКЦИЯ ОПТИМИЗАЦИИ ДЛЯ 3-Х ВХОДОВ
# =====================================================================
def evaluate_3d_fuzzy_system(params):
    # Распаковываем 5 подбираемых параметров (добавился sleep_norm_center)
    hrv_norm, fatigue_med, sleep_norm, load_keep, load_boost = params

    # Защитные барьеры для геометрии
    if not (-20 <= load_keep <= load_boost <= 20): return 1e7
    if not (40 <= hrv_norm <= 70) or not (3 <= fatigue_med <= 7): return 1e7
    if not (6.0 <= sleep_norm <= 8.5): return 1e7 # Норма сна должна быть в разумных пределах

    # Инициализация шкал
    hrv = ctrl.Antecedent(np.arange(0, 101, 1), 'hrv')
    fatigue = ctrl.Antecedent(np.arange(0, 11, 1), 'fatigue')
    sleep = ctrl.Antecedent(np.arange(0, 13, 0.5), 'sleep') # Сон от 0 до 12 часов
    load_mod = ctrl.Consequent(np.arange(-50, 21, 1), 'load_mod')

    # Настройка графиков
    hrv['low'] = fuzz.trapmf(hrv.universe, [0, 0, 30, hrv_norm])
    hrv['normal'] = fuzz.trimf(hrv.universe, [30, hrv_norm, 80])
    hrv['excellent'] = fuzz.trapmf(hrv.universe, [hrv_norm, 80, 100, 100])

    fatigue['low'] = fuzz.trimf(fatigue.universe, [0, 0, fatigue_med])
    fatigue['medium'] = fuzz.trimf(fatigue.universe, [2, fatigue_med, 8])
    fatigue['high'] = fuzz.trapmf(fatigue.universe, [fatigue_med, 8, 10, 10])

    # Новый график сна: плохой, нормальный, отличный
    sleep['bad'] = fuzz.trapmf(sleep.universe, [0, 0, 5.0, sleep_norm])
    sleep['normal'] = fuzz.trimf(sleep.universe, [5.0, sleep_norm, 9.0])
    sleep['good'] = fuzz.trapmf(sleep.universe, [sleep_norm, 9.0, 12, 12])

    load_mod['drop'] = fuzz.trapmf(load_mod.universe, [-50, -50, -25, load_keep])
    load_mod['keep'] = fuzz.trimf(load_mod.universe, [-20, load_keep, load_boost])
    load_mod['boost'] = fuzz.trimf(load_mod.universe, [load_keep, load_boost, 20])

    # Обновленная база правил под 3 параметра
    rule1 = ctrl.Rule(hrv['low'] | fatigue['high'] | sleep['bad'], load_mod['drop'])
    rule2 = ctrl.Rule(hrv['normal'] & fatigue['medium'] & sleep['normal'], load_mod['keep'])
    rule3 = ctrl.Rule(hrv['excellent'] & fatigue['low'] & sleep['good'], load_mod['boost'])

    training_control = ctrl.ControlSystem([rule1, rule2, rule3])
    sim = ctrl.ControlSystemSimulation(training_control)

    total_error = 0
    for row in historical_data:
        sim.input['hrv'] = row[0]
        sim.input['fatigue'] = row[1]
        sim.input['sleep'] = row[2]
        try:
            sim.compute()
            total_error += (sim.output['load_mod'] - row[3]) ** 2
        except:
            total_error += 10000

    return total_error

# =====================================================================
# 3. ЗАПУСК И ОТРЕСОВКА
# =====================================================================
print("Запуск автоподбора параметров (ВСР + Усталость + Сон)...")

# Начальные догадки: [hrv_norm, fatigue_med, sleep_norm, load_keep, load_boost]
initial_guess = [50.0, 5.0, 7.5, 0.0, 10.0]
bounds = [(35.0, 75.0), (3.0, 7.0), (6.0, 8.5), (-15.0, 5.0), (6.0, 18.0)]

result = minimize(evaluate_3d_fuzzy_system, initial_guess, method='L-BFGS-B', bounds=bounds)
h_cent, f_cent, s_cent, l_keep, l_boost = result.x

# Рисуем графики
x_sleep = np.arange(0, 13, 0.5)
plt.figure(figsize=(18, 4))

# График сна (Новый)
plt.subplot(1, 4, 1)
plt.plot(x_sleep, fuzz.trapmf(x_sleep, [0, 0, 5.0, s_cent]), label='Плохой', color='red')
plt.plot(x_sleep, fuzz.trimf(x_sleep, [5.0, s_cent, 9.0]), label='Норма', color='orange')
plt.plot(x_sleep, fuzz.trapmf(x_sleep, [s_cent, 9.0, 12, 12]), label='Отличный', color='green')
plt.axvline(s_cent, color='blue', linestyle='--', alpha=0.7)
plt.title(f'Сон (Идеальный центр: {s_cent:.1f} ч.)')
plt.xlabel('Время сна в часах')
plt.ylabel('Степень уверенности')
plt.legend()

# Остальные графики для наглядности
plt.subplot(1, 4, 2)
plt.plot(np.arange(0, 101, 1), fuzz.trimf(np.arange(0, 101, 1), [30, h_cent, 80]), color='orange', label='Норма ВСР')
plt.axvline(h_cent, color='blue', linestyle='--')
plt.title(f'ВСР Норма: {h_cent:.1f}')
plt.legend()

plt.subplot(1, 4, 3)
plt.plot(np.arange(0, 11, 1), fuzz.trimf(np.arange(0, 11, 1), [2, f_cent, 8]), color='orange', label='Ср. Усталость')
plt.axvline(f_cent, color='blue', linestyle='--')
plt.title(f'Пик Усталости: {f_cent:.1f}')
plt.legend()

plt.subplot(1, 4, 4)
plt.plot(np.arange(-50, 21, 1), fuzz.trimf(np.arange(-50, 21, 1), [-20, l_keep, l_boost]), color='orange', label='Сохранить')
plt.axvline(l_keep, color='blue', linestyle='--')
plt.title(f'Точка сохранения: {l_keep:.1f}%')
plt.legend()

plt.tight_layout()
plt.show()

print(f"\nВывод: Оптимизатор определил, что для данного атлета критический центр 'нормального сна' равен {s_cent:.1f} ч.")
print(f"Если сон падает ниже {s_cent:.1f} ч., кривая 'Плохой сон' начинает стремительно расти вверх, активируя правило сброса нагрузок.")
```
